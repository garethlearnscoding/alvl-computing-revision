```python
# ===== Shared helpers: run this cell first =====
import random

# ---------- Linked list ----------
class ListNode:
    def __init__(self, val, next=None):
        self.val = val
        self.next = next

def build_list(arr):
    head = None
    for x in reversed(arr):
        head = ListNode(x, head)
    return head

def list_to_array(head):
    out = []
    while head:
        out.append(head.val)
        head = head.next
    return out

# ---------- Binary search tree ----------
class TreeNode:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def build_bst(arr):
    """Builds a BST by ITERATIVE insertion (so it does not give away the recursive answer)."""
    root = None
    for x in arr:
        new = TreeNode(x)
        if root is None:
            root = new
            continue
        cur = root
        while True:
            if x < cur.val:
                if cur.left is None:
                    cur.left = new
                    break
                cur = cur.left
            else:
                if cur.right is None:
                    cur.right = new
                    break
                cur = cur.right
    return root

def inorder_iter(root):
    out, stack, cur = [], [], root
    while stack or cur:
        while cur:
            stack.append(cur)
            cur = cur.left
        cur = stack.pop()
        out.append(cur.val)
        cur = cur.right
    return out

# ---------- Queue (array based) ----------
class Queue:
    def __init__(self):
        self._data = []
    def enqueue(self, x):
        self._data.append(x)
    def dequeue(self):
        return self._data.pop(0)
    def peek(self):
        return self._data[0]
    def is_empty(self):
        return len(self._data) == 0
    def size(self):
        return len(self._data)
    def to_list(self):
        return list(self._data)

def make_queue(arr):
    q = Queue()
    for x in arr:
        q.enqueue(x)
    return q

# ---------- Hash table (separate chaining) ----------
class HashTable:
    """Each bucket is the head of a linked list of ListNode(val=(key, value))."""
    def __init__(self, size=5):
        self.size = size
        self.buckets = [None] * size
        self.count = 0

    def _hash(self, key):
        if isinstance(key, int):
            return key % self.size
        return sum(ord(c) for c in key) % self.size

    def put(self, key, value):          # iterative helper, provided for you
        i = self._hash(key)
        node = self.buckets[i]
        while node:
            if node.val[0] == key:
                node.val = (key, value)
                return
            node = node.next
        self.buckets[i] = ListNode((key, value), self.buckets[i])
        self.count += 1

def table_items(t):
    out = []
    for node in t.buckets:
        while node:
            out.append(node.val)
            node = node.next
    return sorted(out)

print("Helpers loaded.")
```
