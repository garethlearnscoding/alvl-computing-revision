```python
from flask import Flask,render_template,request

@app.route('/')
def index():
    return render_template("index.html")

@app.route('/upload',methods=["POST"])
def upload():
    if request:
        username = request.get("username")
        age = request.get("age")
    return render_template("index.html",username=username,age=age)
```
