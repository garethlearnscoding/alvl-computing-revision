## Bubble Sort
1. Loop through the list
2. For each element, compare to the next element, swap if necessary [forwards comparison]
3. **Optimisation:** Tracking if there has been no swaps made and end if there is a swap made

```python
def bubble(arr):
    for i in range(len(arr)):
        swap = False
        for j in range(len(arr)-i-1):
            if arr[j] > arr[j+1]:
                arr[j],arr[j+1] = arr[j+1],arr[j]
                swap = True
        if not swap:
            break

arr = [321,2131,1,43,53,23,0,2,34,2,1,4,532,12]
bubble(arr)
print(arr)
```

## Insertion Sort
1. Loop through the list
2. For each element, compare to the prev element, swap if necessary [backwards comparison]

```python
def insertion(arr):
    for i in range(1,len(arr)):
        while i>0 and arr[i] <   arr[i-1]:
            arr[i],arr[i-1] = arr[i-1],arr[i]
            i-=1

arr = [321,2131,1,43,53,23,0,2,34,2,1,4,532,12]
insertion(arr)
print(arr)
```

## Quick Sort

1. Loop through the list
2. choose a pivot (at random - for optimisation)
3. Create 3 sub list for elements lesser, equal and greater than the pivot
4. recursively call the quicksort function on the lesser and greater sub list

```python
def quick(arr):
    import random

    if len(arr) <= 1:
        return []

    pivot = random.choice(arr)

    lesser = [i for i in arr if i < pivot]
    equal = [i for i in arr if i == pivot]
    greater = [i for i in arr if i > pivot]


    return quick(lesser)+equal+quick(greater)

arr = [321,2131,1,43,53,23,0,2,34,2,1,4,532,12]
arr = quick(arr)
print(arr)
```

## Quick Sort [In Place]

1. loop through the 
2. derive mid using a partition function
3. recursively call the helper function on the elements before mid and those after mid

**Partion 1 [Lomuto]**

1. set pivot to the last element
2. instantiate a pointer at low - 1
3. loop through from low to high
4. if lesser / greater increment i by 1 and swap i and j elements
5. swap pivot to last item greater than pivot [i.e. i+1]
6. return i+1 as midpoint [i.e pivot position]

* Note: `i` tracks the index of the last element smaller than the pivot

**Partion 2 [Hoare's]**

1. Set pivot to the first element
2. instantiate 2 pointers at low and high
3. loop and increment and decrement low and high pointers respectively
4. simulatenously check if low still less than high then flip
5. else return j as midpoint

```python
def lomuto_quick(arr):
    lomuto_helper(arr,0,len(arr)-1)

def lomuto_helper(arr,low,high):
    if low < high:
        mid = lomuto_partion(arr,low,high)
        lomuto_helper(arr,low,mid-1)
        lomuto_helper(arr,mid+1,high)

def lomuto_partion(arr,low,high):
    pivot = arr[high]
    i = low -1
    for j in range(low,high):
        if arr[j] <= pivot:
            i += 1
            arr[j],arr[i] = arr[i],arr[j]
    arr[i+1],arr[high] = arr[high],arr[i+1]
    return i+1



arr = [321,2131,1,43,53,23,0,2,34,2,1,4,532,12]
lomuto_quick(arr)
print(arr)
```

```python
def hoare_quick(arr):
    hoare_helper(arr,0,len(arr)-1)

def hoare_helper(arr,low,high):
    if low < high:
        mid = hoare_partition(arr,low,high)
        hoare_helper(arr,low,mid)
        hoare_helper(arr,mid+1,high)

def hoare_partition(arr,low,high):
    pivot = arr[low]
    i = low -1
    j = high + 1
    while True:
        j-= 1
        while arr[j] > pivot:
            j-=1
        i+=1
        while arr[i] < pivot:
            i += 1
        if i<j:
            arr[i],arr[j] = arr[j],arr[i]
        else:
            return j


arr = [321,2131,1,43,53,23,0,2,34,2,1,4,532,12]
hoare_quick(arr)
print(arr)
```

## Merge Sort

1. choose midpoint of list as pivot
2. split the list and call the function recursively
3. in each function call join both list in correct order

```python
def merge(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr)//2
    left = merge(arr[:mid])
    right = merge(arr[mid:])

    return merge_helper(left,right)

def merge_helper(left,right):
    c = []
    while len(left) >= 1 and len(right) >= 1:
        if left[0] < right[0]:
            c.append(left.pop(0))
        else:
            c.append(right.pop(0))
    c += left + right
    return c


arr = [321,2131,1,43,53,23,0,2,34,2,1,4,532,12]
arr = merge(arr)
print(arr)
```
