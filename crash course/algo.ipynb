## Bubble Sort
1. Loop through the list
2. For each element, compare to the next element, swap if necessary [forwards comparison]
3. **Optimisation:** Tracking if there has been no swaps made and end if there is a swap made

```python
def bubble(arr):
    for i in range(len(arr)):
        swap = False
        for j in range(len(arr)-i-1):
            if arr[j] > arr[j+1]:
                arr[j],arr[j+1] = arr[j+1],arr[j]
                swap = True
        if not swap:
            break

arr = [321,2131,1,43,53,23,0,2,34,2,1,4,532,12]
bubble(arr)
print(arr)
```

## Insertion Sort
1. Loop through the list
2. For each element, compare to the prev element, swap if necessary [backwards comparison]

```python
def insertion(arr):
    for i in range(1,len(arr)):
        while i>0 and arr[i] <   arr[i-1]:
            arr[i],arr[i-1] = arr[i-1],arr[i]
            i-=1

arr = [321,2131,1,43,53,23,0,2,34,2,1,4,532,12]
insertion(arr)
print(arr)
```

## Quick Sort

1. Loop through the list
2. choose a pivot (at random - for optimisation)
3. Create 3 sub list for elements lesser, equal and greater than the pivot
4. recursively call the quicksort function on the lesser and greater sub list

```python
def quick(arr):
    import random

    if len(arr) <= 1:
        return []

    pivot = random.choice(arr)

    lesser = [i for i in arr if i < pivot]
    equal = [i for i in arr if i == pivot]
    greater = [i for i in arr if i > pivot]


    return quick(lesser)+equal+quick(greater)

arr = [321,2131,1,43,53,23,0,2,34,2,1,4,532,12]
arr = quick(arr)
print(arr)
```
