```python
# Linear Search

def linear_search(arr,val):
    for i in range(len(arr)):
        if arr[i] == val:
            return i
```

```python
# Binary Search - Iterative

def binary_search_iter(arr,val):
    lo = 0
    hi = len(arr)-1

    while hi > lo:
        mid = (lo + hi) // 2
        if arr[mid] == val:
            return mid
        elif val > arr[mid]:
            lo = mid + 1
        else:
            hi = mid-1

    return False

```

```python
# Binary Search - Recursive

def binary_search_rec(arr,val,lo=None,hi=None):
    if lo is None and hi is None:
        lo = 0
        hi = len(arr)-1
    if hi < lo:
        return False
    
    mid = (lo+hi) // 2
    if arr[mid] == val:
        return mid
    elif val > arr[mid]:
        return binary_search_rec(arr,val,mid+1,hi)
    else:
        return binary_search_rec(arr,val,lo,mid-1)
```

```python
# Bubble Sort - Optimised

def bubble_sort(arr):
    for i in range(len(arr)-1):
        swap = False
        for j in range(len(arr)-1-i):
            if arr[j] > arr[j+1]:
                arr[j],arr[j+1] = arr[j+1],arr[j]
                swap = True
        if not swap:
            return
```

```python
# Insertion Sort

def insertion_sort(arr):
    for i in range(len(arr)):
        while i > 0 and arr[i-1] > arr[i]:
            arr[i-1],arr[i] = arr[i],arr[i-1]
            i -= 1
```

```python
# Merge Sort

def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr)//2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    return merge(left,right)

def merge(a,b):
    c = []
    i = j = 0

    while len(a) >= 1 and len(b) >= 1:
        if a[i] <= b[j]:
            c.append(a[i])
            i += 1
        else:
            c.append(b[j])
            j += 1

    c = c + a[i:] + b[j:]

    return c
```

```python
# Quick Sort
def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    import random
    pivot = random.choice(arr)

    lesser = [i for i in arr if i<pivot]
    equal = [i for i in arr if i == pivot]
    greater = [i for i in arr if i > pivot]

    return quick_sort(lesser) + equal + quick_sort(greater)
```

```python
# Quick Sort - In Place [Lomuto's]

def quick_sort_in_place(arr):
    return _quick_sort_in_place(arr,0,len(arr)-1)

def _quick_sort_in_place(arr,lo,hi):
    if lo < hi:
        mid = partition_lomuto(arr,lo,hi)
        _quick_sort_in_place(arr,lo,mid-1)
        _quick_sort_in_place(arr,mid+1,hi)

def partition_lomuto(arr,lo,hi):
    pivot = arr[hi]
    i = lo-1
    for j in range(lo,hi):
        if arr[j] < pivot:
            i += 1
            arr[i],arr[j] = arr[j],arr[i]
    arr[hi],arr[i+1] = arr[i+1],arr[hi] # IMPORTANT: Your pivot needs to be in the actual pivot index

    return i+1


arr = [1,2,3,21,2,232,1,35,34531,3242,34563090963]
quick_sort_in_place(arr)
print(arr)
```

```python
# Quick Sort - In Place [Hoare's]

def quick_sort_in_place(arr):
    return _quick_sort_in_place(arr,0,len(arr)-1)

def _quick_sort_in_place(arr,lo,hi):
    if lo < hi:
        mid = partition_hoare(arr,lo,hi)
        _quick_sort_in_place(arr,lo,mid)
        _quick_sort_in_place(arr,mid+1,hi)

def partition_hoare(arr,lo,hi):
    pivot = arr[lo]
    i = lo-1
    j = hi+1
    while i<j:
        i += 1
        while arr[i] < pivot:
            i += 1
        j -= 1
        while arr[j] > pivot:
            j -= 1
        if i<j:
            arr[j],arr[i] = arr[i],arr[j]
    return j 

        


arr = [1,2,3,21,2,232,1,35,34531,3242,34563090963]
quick_sort_in_place(arr)
print(arr)

```
