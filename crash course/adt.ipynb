# Abstract Data Types

## Stack

```python
class Stack:
    def __init__(self,size):
        self.size = size
        self.array = [None]*size
        self.tail = 0

    def is_empty(self):
        return self.tail == self.size
    def is_full(self):
        return self.tail == 0

    def push(self,data):
        if self.is_full():
            return False
        self.array[self.tail] = data
        self.tail += 1
        return True
    
    def pop(self):
        if self.is_empty():
            return False
        self.tail -= 1
        return self.array[self.tail]
    
    def __str__(self):
        return self.array[0:self.tail]
```

# Circular Queue

```python
class CQueue:
    def __init__(self,size):
        self.array = [None]*size
        self.max_size = size
        self.size = 0
        self.tail = 0
        self.head = 0

    def is_empty(self):
        return self.size == 0

    def is_full(self):
        return self.size == self.max_size
    
    def enqueue(self,data):
        self.array[self.tail] = data
        self.tail = (self.tail + 1)%self.max_size

    def dequeue(self):
        pop = self.array[self.head]
        self.head = (self.head + 1)%self.max_size
        return pop

    def __str__(self):
        lst = []
        curr = self.head
        while curr != self.tail:
            lst.append(self.array[curr])
            curr = (curr+1)%self.max_size
        return lst
```

# Hash Table

```python
class HashTableLinear:
    def __init__(self,size):
        self.size = size
        self.buckets = [None]*size

    def is_full(self):
        return all([(i==None) for i in self.buckets])

    def _hash(self,data):
        return hash(data[1])%self.size

    def insert_linear(self,data):
        idx = self._hash(data)
        while self.buckets[idx] is not None:
            idx = (idx + 1)%self.size
        self.buckets[idx] = data

    def search_linear(self,data):
        idx = self._hash(data)
        j = idx
        while self.buckets[idx] is not None:
            j = (j+1)%self.size
            if j == idx:
                return False
            else:
                return self.buckets[j]

class HashTableChain:
    def __init__(self,size):
        self.size = size
        self.buckets = [None]*size

    def _hash(self,data):
        return hash(data)%self.size

    def insert_chain(self,data):
        idx = self._hash(data)
        curr = self.buckets[idx]
        if curr is None:
            self.buckets[idx] = [data]
        else:
            self.buckets[idx].append(data)

    def search_chain(self,key):
        idx = self._hash(key)
        while self.buckets[idx] is not None:
            for i in self.buckets[idx]:
                if i[0] == key:
                    return i
            return False
        return False
```
