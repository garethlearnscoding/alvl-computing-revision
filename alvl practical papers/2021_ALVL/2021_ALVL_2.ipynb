The IMEI number is a unique value used to identify mobile devices, such as phones and tablets.
The IMEI number has 15 digits: 14 digits with an extra check digit added to the right-hand side.

The check digit is calculated using the following algorithm, on the left-most 14 digits of an IMEI number:
1. starting from the right, the first digit is location number 1
2. double all digits in the odd numbered positions
3. sum all the digits, including both the unchanged digits (i.e. those in the even numbered positions) as well as those doubled (e.g. 16 contributes 1 + 6)
4. the check digit is the value between O and 9 that must be added to the sum to make the esult exactly divisible by 10.

For example, given the 14 digits 14576567654934:

<center>

||||||||||||||||
|---|---|---|---|---|---|---|---|---|---|---|---|---|---|---|
| Step 1: | 1 | 4 | 5 | 7 | 6 | 5 | 6 | 7 | 6 | 5 | 4 | 9 | 3 | 4 |
| Step 2: | 1 | 8 | 5 | 14 | 6 | 10 | 6 | 14 | 6 | 10 | 4 | 18 | 3 | 8 |
| Step 3: | 1+ | (8)+ | 5+ | (1+4)+ | 6+ | (1+0)+ | 6+ | (1+4)+ | 6+ | (1+0)+ | 4 | (1+8)+ | 3 | (8) |

</center>

Sum= 68

Step 4: Check digit = 2


## Task 1.1
Write a function `task1_1(input_value)` that returns an integer.
The function should:
- validate that the parameter `input_value` is either an integer, or a string containing a valid integer
- check that it is 14 digits in length
- return -1 if the value received is invalid for any reason
- calculate the check digit for the given first 14 digits of the IMEI number
- return the calculated check digit. 

<div style="text-align: right">[6]</div> 

````python
# [1] validates input is int OR string containing digits only
# [1] checks that the input has exactly 14 digits
# [1] correctly processes digits from right to left / applies correct alternating doubling
# [1] correctly handles doubled values greater than 9 by summing their digits
# [1] correctly totals all processed digits
# [1] correctly calculates and returns the check digit, i.e. (10 - total % 10) % 10

def valid_input_14(input_value):
    if (type(input_value)==int or (type(input_value)==str and input_value.isdigit())) and (len(str(input_value))==14):
        return True
    print('Input value has to be 14 digits long and each digit has to be an integer. Please try again')
    return False

def digit_sum(n):
    n=str(n)
    return sum([int(i) for i in n])

def task1_1(input_value):
    if valid_input_14(input_value):
        input_value=str(input_value)
        input_value=[int(i) for i in input_value]
        weights=[1,2]*7
        pairwise_product=[input_value[i]*weights[i] for i in range(len(input_value))]
        pairwise_product=[digit_sum(i) for i in pairwise_product]
        return (10 - (sum(pairwise_product) % 10)) % 10
    return -1
````

````python
# [1] iterates through list from first to last element (outer loop structure)
# [1] uses secondary index to track position for insertion
# [1] compares current element with previous elements (insertion condition)
# [1] correctly swaps/shifts elements to maintain order
# [1] returns sorted list in ascending order
# [1] correct test case returned

def task2_2(arr):
    for i in range(0,len(arr)):
        j = i
        while j > 0 and arr[j]<arr[j-1]:
            arr[j],arr[j-1]=arr[j-1],arr[j]
            j=j-1
    return arr

print(task2_2(task2_1('TEN.txt')))
````

A programmer is writing a class, `LinkedList`, to represent a linked list of unique integers. A linked list is a collection of data elements, whose order is not given by their physical placement in memory. Instead, each element points to the next.

## Task 3.1
Write the `LinkedList` class in Python. Use of a simple Python list is not sufficient. Include the following methods:
- `insert(integer_value)` inserts the `integer_value` at the beginning (head) of the list
- `delete(integer_value)` attempts to delete `integer_value` from the list; if the item was not present, return `None`
- `search(integer_value)` returns a Boolean value: `True` if `integer_value` is in the list, `False` if not in the list
- `count()` should return the number of elements in the list, or zero if empty 
- `to_String()` should return a string containing a suitably formatted list with the elements separated by a comma and a space, with square brackets at either end, e.g. in the form:
    ```
    [11, 2, 7, 4]
    ```
<div style="text-align: right">[8]</div> 

Test `LinkedList` by using the data in the file `Task3data. txt`. Use the `to_String()` method to print the resulting contents of the list. 
<div style="text-align: right">[3]</div> 

````python
# [1] defines Node class/object structure with data and pointer/reference
# [1] LinkedList does not rely on Python list as main storage
# [1] insert adds new item at head correctly
# [1] delete removes head node correctly
# [1] delete removes middle/end node correctly
# [1] search returns True/False correctly
# [1] count returns correct number of elements
# [1] to_String returns correctly formatted string e.g. "[11, 2, 7, 4]"
# [1] reads integers from Task3data.txt
# [1] inserts all values into linked list
# [1] prints/displays resulting list using to_String()

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class LinkedList:
    def __init__(self):
        self.first = None
        self.ct = 0

    def is_empty(self):
        if self.first:
            return False
        else:
            return True

    def insert(self, data):
        new_node = Node(data)
        new_node.next = self.first
        self.first = new_node
        self.ct += 1

    def search(self, data):
        current = self.first
        while current:
            if current.data == data:
                return True
            current = current.next
        return False  

    def delete(self, data):
        if not self.first:
            return

        if self.first.data == data:
            self.first = self.first.next
            self.ct -= 1
            return

        current = self.first
        previous = None

        while current and current.data != data:
            previous = current
            current = current.next

        if not current:
            return  # Value not found
        
        self.ct -= 1

        previous.next = current.next

    def count(self):
        return self.ct

    def to_String(self):
        current = self.first
        to_return = []
        while current:
            to_return.append(current.data)
            current = current.next
        return to_return
````

Write a Python subclass `SortedLinkedList` using `LinkedList` as its superclass.

The insert method in the `SortedLinkedList` subclass should ensure that the elements are stored in ascending order. 
<div style="text-align: right">[5]</div> 

Test `SortedLinkedList` by using the data in the file `Task3data.txt`. Use the `to_String()` method to print the resulting contents of the list.

Print the result of searching the `SortedLinkedList` for the value 94.
<div style="text-align: right">[2]</div> 

````python
# [1] correctly inherits from LinkedList
# [1] handles insertion into empty list
# [1] handles insertion before current head
# [1] handles insertion in middle/end position
# [1] maintains ascending order after multiple insertions
# [1] reads Task3data.txt and displays sorted list using to_String()
# [1] searches for value 94 and prints result

class SortedLinkedList(LinkedList):
    def __init__(self):
        super().__init__()

    def insert(self, data):
        new_node = Node(data)

        # Case 1: empty list or insert before first
        if self.first is None or data < self.first.data:
            new_node.next = self.first
            self.first = new_node
            self.ct += 1
            return

        # Case 2: find position
        current = self.first
        while current.next and current.next.data < data:
            current = current.next

        # insert after current
        new_node.next = current.next
        current.next = new_node
        self.ct += 1
````

Write a Python subclass `Queue` using `LinkedList` as its superclass.

Additional `enqueue` and `dequeue` methods are to be defined on the `Queue` class:
- `enqueue(integer_value)` will insert `integer_value` to the end of the queue
- `dequeue()` will return the first element in the queue. If the queue is empty, return `None`. 
<div style="text-align: right">[6]</div> 

Test `Queue` by using the data in the file `Task3data.txt`. Print the first five elements to be dequeued from the list. <div style="text-align: right">[1]</div> 

````python
# [1] correctly inherits from LinkedList
# [1] defines enqueue method
# [1] enqueue inserts at rear of queue
# [1] defines dequeue method
# [1] dequeue removes and returns front item
# [1] dequeue returns None when queue is empty
# [1] enqueues data from Task3data.txt and prints first five dequeued values

class Queue(LinkedList):
    def __init__(self):
        super().__init__()

    def enqueue(self,data):
        self.insert(data)

    def dequeue(self):
        # Case 1: empty list
        if self.first == None:
            return None # Queue is empty
        
        # Case 2: only one node
        if self.first.next is None:
            value = self.first.data
            self.first = None
            self.ct -= 1
            return value

        # Case 3: more than one node - traverse to second last mode
        previous = self.first
        current = self.first.next

        while current.next:
            previous = current
            current = current.next

        # current is last, previous is second last
        value = current.data
        previous.next = None
        self.ct -= 1
        return value
    
with open('../resources/Task3data.txt') as f:
    data = f.readlines()
    data = [int(i.strip()) for i in data]

q = Queue()

for i in data:
    q.enqueue(i)

for _ in range(5):
    print(q.dequeue())
````
