```python
# Task 1.1
import random

colours = ['red', 'green', 'blue']
num = [1,2,3,4]

player_one = [[random.choice(colours), random.choice(num)] for _ in range(10)]
player_two = [[random.choice(colours), random.choice(num)] for _ in range(10)]

player_one = sorted(player_one, key = lambda x: x[1])
player_two = sorted(player_two, key = lambda x: x[1])

print(player_one)
print(player_two)
```

```python
# Task 1.2
def compare(col_one, col_two):
    if (col_one == 'red' and col_two == 'green') or (col_one == 'green' and col_two == 'blue') or (col_one == 'blue' and col_two == 'red'):
        return '1'
    elif (col_two == 'red' and col_one == 'green') or (col_two == 'green' and col_one == 'blue') or (col_two == 'blue' and col_one == 'red'):
        return '2'
    else:
        return '3'
    
for i in ['red','green','blue']:
    for j in ['red','green','blue']:
        print(i,j, compare(i,j))
```

```python
# FUNCTION 

def insertion_sort_no(lst):
    colour_order = ['red', 'green', 'blue']

    for i in range(1,len(lst)):
            j = i
            while j > 0:
                if lst[j][1] < lst[j-1][1] or ((lst[j][1] == lst[j-1][1]) and (colour_order.index(lst[j][0]) < colour_order.index(lst[j-1][0]))):
                  lst[j], lst[j-1] = lst[j-1], lst[j]
                j = j - 1
    return lst

player_one = [[random.choice(colours), random.choice(num)] for _ in range(10)]

print(player_one)
player_one = insertion_sort_no(player_one)
print(player_one)

player_two = [[random.choice(colours), random.choice(num)] for _ in range(10)]

print(player_two)
player_two = insertion_sort_no(player_two)
print(player_two)
```

```python
# Task 1.4
def calculate_score(player_one,player_two):
    score_one = 0
    score_two = 0 
    for _ in range(10):
        winner = compare(player_one[_][0], player_two[_][0])
        score_one += (2*player_one[_][1] if winner == '1' else player_one[_][1] )
        score_two += (2*player_two[_][1] if winner == '2' else player_two[_][1] )

    print(f'Player 1 score: {score_one}')
    print(f'Player 2 score: {score_two}')

    return score_one, score_two
calculate_score(player_one,player_two)
```

```python
# Task 1.5
def winner(player_one, player_two):
    score_one, score_two = calculate_score(player_one,player_two)
    if score_one > score_two:
        print(f'Player 1 wins with {score_one} pt. Player 2 came 2nd place with {score_two} pt')
    elif score_two > score_one:
        print(f'Player 2 winswith {score_one} pt. Player 1 came 2nd place  with {score_two} pt')
    else:
        print(f'There is a draw as the scores are tied at {score_one}')
```

```python
# Task 2.1
class Node:
    def __init__(self, data_value = -1, left_node = -1, right_node = -1):
        self.data_value = data_value
        self.left_node = left_node
        self.right_node = right_node

    def set_data(self, val):
        self.data_value = val

    def get_data(self):
        return self.data_value
    
    def set_left(self, val):
        self.left_node = val

    def get_left(self):
        return self.left_node
    
    def set_right(self, val):
        self.right_node = val

    def get_right(self):
        return self.right_node
```

```python
# Task 2.2
def calculate_nodes(num):
    print(f'A {num} level number tree has {2**num - 1} nodes')
    return 2**num - 1
```

```python
# Task 3.1
def heading(text):
    return f'<h1>{text}</h1>'

def subheading(text):
    return f'<h2>{text}</h2>'

def para(text):
    return f'<p>{text}</p>'

text = 'My element'

print(heading(text))
print(subheading(text))
print(para(text))
```

```python
# Task 3.2
def hyperlink(text):
    url, link_text = text.split(',')
    return f'<a href = "{url}">{link_text}</a>'

print(hyperlink('https://www.cambridgeinternational.org, Click here'))
```

```python
# Task 3.3
def body(lst):
    html_content = ''

    html_content += '<html>\n'

    html_content += '<body>\n'

    for i in lst:
        if i[0] == 'heading':
            html_content += (heading(i[1])) + '\n'
        elif i[0] == 'subheading':
            html_content += (subheading(i[1])) + '\n'
        elif i[0] == 'paragraph':
            html_content += (para(i[1])) + '\n'
        elif i[0] == 'hyperlink':
            html_content += (hyperlink(''.join(i[1:]))) + '\n'
        else:
            return False

    html_content += '</body>\n'

    html_content += '</html>'

    return html_content
```

```python
# Task 3.4
def read_data(filename):
    with open(f'./resources/{filename}') as f:
        lst = f.readlines()
        lst = [i.strip() for i in lst]
        lst = [i.split(':') for i in lst]
    
    return body(lst)

print(read_data('firstPage.txt'))

def write_data(filename):
    html_content = '<!doctype html>\n'
    if read_data(filename):
        html_content += read_data(filename)
        with open(f'./resources/{filename.split(".")[0]}.html','w') as f:
            f.write(html_content)
    else:
        return f'{filename} contains unrecognised elements'
    
write_data('firstPage.txt')
```
