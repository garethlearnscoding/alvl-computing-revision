````python
# Task 1.4 [6]
# [1] Defines a function to calculate the scores of both players
# [1] Correctly compares corresponding cards from both hands
# [1] Correctly awards double the winning card number to the winning player
# [1] Correctly awards the losing card number to the losing player
# [1] Correctly handles the case where both cards have the same colour
# [1] Outputs each sorted hand and the final score on the same line for both players

# [1] Defines a function to calculate the scores of both players
# The function accepts both players' hands as parameters
def calculate_score(player_one, player_two):

    # Initialises Player 1's score to 0
    score_one = 0

    # Initialises Player 2's score to 0
    score_two = 0 

    # [1] Compares corresponding cards from both hands
    # There are 10 cards in each hand, so the loop runs 10 times
    for _ in range(10):

        # Compares the colour of Player 1's card with the colour of Player 2's card
        # winner will be:
        # '1' if Player 1's colour wins
        # '2' if Player 2's colour wins
        # '3' if both colours are the same
        winner = compare(player_one[_][0], player_two[_][0])

        # [1] Correctly awards double the winning card number to Player 1 if Player 1 wins
        # [1] Correctly awards the normal card number to Player 1 if Player 1 loses or draws
        # This also handles the same-colour case by awarding the normal card number
        score_one += (2 * player_one[_][1] if winner == '1' else player_one[_][1])

        # [1] Correctly awards double the winning card number to Player 2 if Player 2 wins
        # [1] Correctly awards the normal card number to Player 2 if Player 2 loses or draws
        # This also handles the same-colour case by awarding the normal card number
        score_two += (2 * player_two[_][1] if winner == '2' else player_two[_][1])

    # [1] Outputs Player 1's final score and sorted hand on the same line
    print(f'Player 1 score: {score_one} with hand {player_one}')

    # [1] Outputs Player 2's final score and sorted hand on the same line
    print(f'Player 2 score: {score_two} with hand {player_two}')

    # Returns both scores so they can be used later to determine the winner
    return score_one, score_two


# Tests the score calculation function using both players' hands
calculate_score(player_one, player_two)
````

The data for each parent node in the number tree is calculated by adding the data from its two lower nodes.

The function `create_number_tree()`:
- takes 2 parameters: the list of nodes (with the first level of nodes already assigned their data) and the number of levels
- calculates the data for each node in each level in the number tree
- stores the data in each node
- assigns the left pointer of each node to its left lower node
- assigns the right pointer of each node to its right lower node.

Write the function `create_number_tree()`. <div style="text-align: right">[4]</div>

````python
# Task 2.6 [4]
# [1] Correctly processes each non-leaf level of the tree
# [1] Correctly calculates each parent node value by adding its two child node values
# [1] Correctly assigns the left pointer of each parent node
# [1] Correctly assigns the right pointer of each parent node

# Defines a function to create the number tree from the leaf nodes upwards
def create_number_tree(arr, level):

    # [1] Correctly processes each non-leaf level of the tree
    # p represents the current pair of levels being processed, starting from the leaf level
    for p in range(1, level):

        # Calculates the starting index of the lower level currently being processed
        lower_level_bottom = sum([2**(level-i) for i in range(1, p)])

        # Calculates the ending index of the lower level currently being processed
        lower_level_top = sum([2**(level-i) for i in range(1, p+1)])

        # The upper level starts immediately after the lower level
        upper_level_bottom = lower_level_top

        # Calculates the ending index of the upper level currently being filled
        upper_level_top = sum([2**(level-i) for i in range(1, p+2)])

        # Stores the left child index, right child index and calculated parent value
        to_add = []

        # Processes the lower level two nodes at a time
        for i in range(lower_level_bottom, lower_level_top, 2):

            # [1] Correctly calculates each parent node value
            # by adding the data values of its two child nodes
            to_add.append((i, i+1, arr[i].get_data() + arr[i+1].get_data()))

        # Assigns the calculated parent values and pointers to the upper level nodes
        for j in range(upper_level_bottom, upper_level_top):

            # Sets the parent node data value
            arr[j].set_data(to_add[j - lower_level_top][2])

            # [1] Correctly assigns the left pointer of each parent node
            arr[j].set_left(to_add[j - lower_level_top][0])

            # [1] Correctly assigns the right pointer of each parent node
            arr[j].set_right(to_add[j - lower_level_top][1])
        
    # Returns the completed number tree
    return arr
````

## Task 3.3
The function `body()` takes a list as a parameter containing the text to convert into an HTML formatted document. Each element in the list has one HTML element for conversion.

Write the function to:
- call the appropriate function for each HTML element in the list
- create a single string containing all the formatted returned HTML elements
- insert the HTML and body opening and closing tags at the start and end of the string
- return either the complete formatted string, or False if there are any unrecognised elements. <div style="text-align: right">[6]</div>

````python
# Task 3.3 [6]
# [1] Iterates through each element in the input list
# [1] Correctly identifies and processes heading elements
# [1] Correctly identifies and processes subheading elements
# [1] Correctly identifies and processes paragraph elements
# [1] Correctly identifies and processes hyperlink elements
# [1] Adds opening and closing HTML/body tags and returns False for unrecognised elements

# Defines a function to create the HTML body from a list of elements
def body(lst):

    # Initialises the HTML content as an empty string
    html_content = ''

    # [1] Adds the opening HTML tag
    html_content += '<html>\n'

    # [1] Adds the opening body tag
    html_content += '<body>\n'

    # [1] Iterates through each element in the input list
    for i in lst:

        # [1] Correctly identifies and processes heading elements
        # i[1] contains the text to be formatted as a heading
        if i[0] == 'heading':
            html_content += (heading(i[1])) + '\n'

        # [1] Correctly identifies and processes subheading elements
        # i[1] contains the text to be formatted as a subheading
        elif i[0] == 'subheading':
            html_content += (subheading(i[1])) + '\n'

        # [1] Correctly identifies and processes paragraph elements
        # i[1] contains the text to be formatted as a paragraph
        elif i[0] == 'paragraph':
            html_content += (para(i[1])) + '\n'

        # [1] Correctly identifies and processes hyperlink elements
        # ''.join(i[1:]) is used so that the remaining parts form the hyperlink text
        elif i[0] == 'hyperlink':
            html_content += (hyperlink(''.join(i[1:]))) + '\n'

        # [1] Returns False if the element type is not recognised
        else:
            return False

    # [1] Adds the closing body tag
    html_content += '</body>\n'

    # [1] Adds the closing HTML tag
    html_content += '</html>'

    # Returns the completed HTML string
    return html_content
````

````python
# Task 4.4 [7]
# [1] Function accepts villa name, start date and number of days
# [1] Correctly generates all requested dates for the booking period
# [1] Uses SQL to find existing booked dates for the given villa
# [1] Correctly identifies dates that are available
# [1] Correctly identifies dates that are unavailable
# [1] Outputs both available and unavailable dates with appropriate messages
# [1] Tests the function using Dolphin, Apr 8 and 4 days

# Helper function to generate all dates to be checked for the requested booking period
def dates_to_check(start_date, no_of_days):

    # [1] Correctly generates all requested dates for the booking period
    # For no_of_days = n, this generates dates from day 0 to day n - 1
    return [add_date(start_date, j) for j in range(no_of_days)]


# [1] Defines a function that accepts villa name, start date and number of days
def booking_check(villa_name, start_date, no_of_days):

    import sqlite3

    # Connects to the SQLite database
    connection = sqlite3.connect("holiday_company.db")

    # [1] Uses SQL to find existing booked dates for the given villa
    # Joins Villa_Booking and Villa so that the search can be done using VillaName
    booking_info = connection.execute(f'''SELECT VillaName, DateBooked
        FROM Villa_Booking
        INNER JOIN Villa
        ON Villa_Booking.VillaID = Villa.VillaID
        WHERE VillaName = "{villa_name}"
        ''')

    # Extracts the booked dates for the selected villa into a list
    booked_dates = [i[1] for i in list(booking_info)]
    
    # Stores dates that are available
    available = []

    # Stores dates that are already booked
    not_available = []

    # Checks each requested date against the booked dates from the database
    for date in dates_to_check(start_date, no_of_days):

        # [1] Correctly identifies dates that are unavailable
        if date in booked_dates:
            not_available.append(date)

        # [1] Correctly identifies dates that are available
        else:
            available.append(date)

    # [1] Outputs the available dates with an appropriate heading
    print('---------------')
    print('AVAILABLE DATES')
    print('---------------')
    for date in available:
        print(date)

    # [1] Outputs the unavailable dates with an appropriate heading
    print('---------------')
    print('UNAVAILABLE DATES')
    print('---------------')
    for date in not_available:
        print(date)


# [1] This tests Dolphin and Apr 8 for 4 days
booking_check('Dolphin', '08-APR', 4)
````
