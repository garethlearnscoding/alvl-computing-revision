The task is to implement a hashing function using the modulus function and ASCll codes.
The hash is implemented with the following pseudocode, acting on a string `string_value`, which returns an integer. `h`, representing the hash.

```text
h <- 0
FOR i <- 0 TO length(string_value) - 1
    val <- 33 * (ASCII value of string_value[i])
    h <- (h + val) % 1024
NEXT i
RETURN h
```

## Task 1.1

Write a function `task1_1(string_value)` that:
- takes a string value `string_value`
- implements the hash algorithm to produce an integer value
- returns that integer value. 
<div style="text-align: right">[5]</div> 

Test your function using the following **three** calls:
```text
task1_1("Hello")
task1_1("Hallo")
task1_1("Hullo") 
```
<div style="text-align: right">[3]</div> 

````python
# [1] Iterates through every character in string_value
# [1] Uses ord() / ASCII value of each character
# [1] Multiplies ASCII value by 33
# [1] Updates hash using (h + val) % 1024 
# [1] correct final value returned

# [1] correct return for "Hello"
# [1] correct return for "Hallo"
# [1] correct return for "Hullo"

def task1_1(string_value):
    h = 0 
    for i in range(len(string_value)):
        val = 33 * ord(string_value[i])
        h = (h + val) % 1024
    return h

test_cases = ["Hello", "Hallo", "Hullo"]

for t in test_cases:
    print(task1_1(t))
````

Write a function `task2_2(list_of_integers)` that:

<div style="text-align: right">[2]</div> 

````python
# [1] Correct base case for list of length 0 or 1
# [1] Splits list into two halves
# [1] Recursively sorts left half
# [1] Recursively sorts right half
# [1] Correctly compares values from both halves
# [1] Correctly appends remaining values after one half is exhausted
# [1] Returns fully sorted list in ascending order

# [1] Uses the given test list
# [1] Correctly checks against expected sorted list / prints True
def merge(A, B):

    C = []

    while (len(A)>=1 and len(B)>=1):
        if A[0] >= B[0]:
            C.append(B[0])
            B.pop(0)
        else:
            C.append(A[0])
            A.pop(0)

    while len(A) >= 1:
        C.append(A[0])
        A.pop(0)

    while len(B) >= 1:
        C.append(B[0])
        B.pop(0)

    return C

def task2_2(A):
    if len(A) <= 1:
        return A

    left = A[:len(A)//2]
    right = A[len(A)//2:]

    left = task2_2(left)
    right = task2_2(right)

    return merge(left, right)


print(task2_2 ([56, 25, 4, 98, 0, 18, 4, 5, 7, 0]) == [0, 0, 4, 4, 5, 7, 18,25, 56, 98])

print(task2_2 ([56, 25, 4, 98, 0, 18, 41, 5, 7, 0]))
````

The task is to write a function that takes a sequence of characters that represents a quantity of data and unit. and translates this quantity to a different unit.

The basic unit of data is the byte (B):

- A kilobyte (KB) is $10^3$ bytes
- A megabyte (MB) is $10^6$ bytes
- A gigabyte (GB) is $10^9$ bytes
- A terabyte (TB) is $10^{12}$ bytes.

For example. 8KB has 8000 bytes.

## Task 3.1
Write a function called `task3_1(quantity_of_data)` that:
- takes a string, `quantity_of_data`
- tests that the given string is a sequence of digits followed by one of the four approved units shown above (KB, MB, GB. TB).
- returns and displays either:
    - the actual number of bytes represented by the input string
    or
    - the error message. `"invalid data"`. 
<div style="text-align: right">[5]</div> 

Test the function fully with suitable test data, including all four approved units.

For example,

`task3_1("8KB")`

should return and display `8000` 
<div style="text-align: right">[3]</div> 


````python
# [1] Extracts and validates unit as one of KB, MB, GB, TB
# [1] Checks quantity part contains only digits
# [1] Uses correct powers of 10: 10^3, 10^6, 10^9, 10^12
# [1] check for "invalid data" for invalid input
# [1] correct value returned

# [1] Includes valid test for KB
# [1] Includes valid tests covering MB/GB/TB
# [1] Includes invalid test data
def task3_1(quantity_of_data):
    if quantity_of_data[-2:] not in ['KB','MB','GB','TB']:
        return  "invalid data"
    else:
        if quantity_of_data[:-2].isdigit():
            factor = {'KB':10**3,
                      'MB':10**6,
                      'GB':10**9,
                      'TB':10**12
                      }
            return int(quantity_of_data[:-2])*factor[quantity_of_data[-2:]]
        else:
            return "invalid data"
        
test_cases = ['AKB48','8KB','16MB','32GB','64TB']

for t in test_cases:
    print(task3_1(t))
````

The school wants to use the Python programming language and object-oriented programming to help publish the database content on a web page.

The class `Person` will store the following data:

- `full_name` - stored as a string
- `date_of_birth` - initialised with a string with the format YYYY-MM-DD

The class has two methods defined on it:
- `is_adult()` - returns a Boolean value to indicate whether the person is an adult or not. It:
    - subtracts the year of the `date_of_birth` from the year of today‘s date
    - returns True if the result is greater than 18, otherwise returns False.
- `screen_name()` - returns a string which creates an identifier to be used as a screen name, which should be constructed as follows:
    - the full name with all spaces and punctuation removed
    - followed by the two digit month of their birth
    - then the two-digit day of their birth.

For example, John Tan, born on the 1st of June 2000 (`"2000-06-01"`), would have the screen name `"JohnTan0601"`

Save your program code as

````python
# [1] Defines Person class correctly
# [1] Constructor stores full_name
# [1] Constructor stores date_of_birth
# [1] is_adult() extracts birth year and compares with current year
# [1] Returns True only when age difference is greater than 18
# [1] screen_name() removes spaces and punctuation from name
# [1] Appends birth month and day in MMDD format
class Person:
    def __init__(self, full_name, date_of_birth):
        self.full_name = full_name
        self.date_of_birth = date_of_birth

    def get_name(self):
        return self.full_name
    
    def set_name(self, full_name):
        self.full_name = full_name

    def get_dob(self):
        return self.date_of_birth
    
    def set_dob(self, date_of_birth):
        self.date_of_birth = date_of_birth

    def is_adult(self):
        from datetime import date
        
        current_year = date.today().year
        
        if (current_year - int(self.get_dob()[:4])) > 18:
            return 1
        else:
            return 0  
        
    def screen_name(self):
        cleaned_name = ''.join([i for i in self.get_name() if i.isalpha()])
        cleaned_date = ''.join(self.get_dob()[-5:].split('-'))
        return cleaned_name + cleaned_date
    
JT = Person("John Tan's", '2000-06-01')
print(JT.screen_name())
````

````python
# [1] Staff inherits from Person
# [1] Staff.screen_name() appends "Staff"
# [1] Staff.is_adult() always returns True
# [1] Student.is_adult() always returns False
class Staff(Person):
    def __init__(self, full_name, date_of_birth):
        super().__init__(full_name, date_of_birth)

    def is_adult(self):
        return 1
    
    def screen_name(self):
        return super().screen_name() + 'Staff'
    
class Student(Person):
    def __init__(self, full_name, date_of_birth):
        super().__init__(full_name, date_of_birth)

    def is_adult(self):
        return 0

    
JTStaff = Staff("John Tan's", '2000-06-01')
print(JTStaff.screen_name())

JTStudent = Student("John Tan's", '2000-06-01')
print(JTStudent.screen_name())
````

````python
# [1] Opens and reads people.txt
# [1] Splits each line into name, date of birth and type
# [1] Creates Person, Staff or Student object appropriately
# [1] Stores created objects for later database insertion
with open('./resources/people.txt') as f:
    data = f.readlines()
    data = [d.strip() for d in data]
    print(data)

people_list = []

for d in data:
    name, dob, c_type = d.split(',')
    if c_type == 'Person':
        people_list.append(Person(name,dob))
    elif c_type == 'Staff':
        people_list.append(Staff(name,dob))
    elif c_type == 'Student':
        people_list.append(Student(name,dob))

for p in people_list:
    print(p.get_name(), p.get_dob(), p.screen_name(), p.is_adult())
````
