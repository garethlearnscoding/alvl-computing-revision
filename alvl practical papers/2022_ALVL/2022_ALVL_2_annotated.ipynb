A Caesar cypher encodes each letter with a different letter. A 10-place Caesar cypher uses the ASCII value of each letter and adds the number 10 to it.
For example:
- The letter 'A' has the ASCII value 65. Adding 10 to the number will give the ASCII value 75. The character for 75 is 'K'.
- When an uppercase letter's code goes beyond 'Z' it returns to 'A'. For example, the character 'Z' will be encrypted as 'J'.
- When a lowercase letter's code goes beyond 'z' it returns to 'a'. For example, the character 'x' will be encrypted as 'h'.
- Spaces ' ' are replaced with the character'!'.

You will only need to convert letters and spaces. If the character is invalid, -1 is returned.

The following table shows the ASCII values of some of the characters.

<center>

| Character | ASCII value |
|---|---|
| A | 65 |
| Z | 90 |
| a | 97 |
| z | 122 |
| ''(space) | 32 |
| ! | 33 |

</center>

In ASCII the letters follow on numerically. For example, the letter 'A' is 65, 'B' is 66, 'C' is 67 etc.

## Task 1
This program will encrypt each letter by adding the number 10 to its ASCII value. Spaces will be replaced with '!'. If the character is invalid, -1 is returned.

Write program code for a function that takes a character as a parameter and returns the ASCII value of the new encrypted character. <div style="text-align: right">[5]</div> 

````python
# Marking scheme
# [1] Correctly identifies uppercase letters A-Z
# [1] Correctly shifts uppercase letters by 10, including wrap-around after Z
# [1] Correctly identifies lowercase letters a-z
# [1] Correctly shifts lowercase letters by 10, including wrap-around after z
# [1] Correctly maps space to '!' and returns -1 for invalid characters

# Candidate comments
def encrypt_ascii(char):
    # Encrypt a single character according to the required ASCII shift rules.
    if char.isupper():
        # Handle uppercase letters first.
        if ord(char) + 10 <= 90:
            # No wrap-around needed for uppercase.
            return chr(ord(char) + 10)
        else:
            # Wrap around from 'Z' back to 'A' for uppercase.
            return chr( 65 + (ord(char) + 10) % 90 - 1 )
        
    if char.islower():
        # Handle lowercase letters similarly.
        if ord(char) + 10 <= 122:
            # No wrap-around needed for lowercase.
            return chr(ord(char) + 10)
        else:
            # Wrap around from 'Z' back to 'A' for uppercase.
            return chr( 97 + (ord(char) + 10) % 122 -1 )
    elif char == ' ':
        # Special case: space is converted to '!'.
        return '!'
    # Any other character is treated as invalid.
    return -1

print(encrypt_ascii('Z'))
print(encrypt_ascii('a'))
print(encrypt_ascii('!'))
print(encrypt_ascii(' '))
````

````python
# Marking scheme
# Class and constructor [4]
# [1] Declares class Tree
# [1] Includes left_pointer, right_pointer and data attributes
# [1] Constructor sets left and right pointers to None
# [1] Constructor stores data parameter correctly

# Insert method [6]
# [1] Declares recursive insert method with new value as parameter
# [1] Compares new value with current node data
# [1] Moves left for smaller values
# [1] Moves right for larger values
# [1] Creates a new Tree node when correct empty position is found
# [1] Correctly handles repeated recursive insertion without losing existing nodes

# Main program [5]
# [1] Creates a new Tree instance
# [1] Generates random integers between 0 and 999 inclusive
# [1] Ensures 10 unique values are generated
# [1] Inserts each unique value into the tree
# [1] Does not use built-in sorting to create the tree order

# Traversals [7]
# [1] Declares in-order traversal method
# [1] In-order traversal visits left subtree
# [1] In-order traversal visits current node
# [1] In-order traversal visits right subtree
# [1] Declares post-order traversal method
# [1] Post-order traversal visits left, right, then current node
# [1] Calls both traversal methods using the tree

# Testing [2]
# [1] Shows in-order traversal output
# [1] Shows post-order traversal output

# Candidate comments
import random

class Tree:
    # Binary search tree node / tree structure.
    def __init__(self, data = None):
        # Initialise pointers and stored data.
        self.left_pointer = None
        self.right_pointer = None
        self.data = data

    def get_left(self):
        return self.left_pointer
    
    def get_right(self):
        return self.right_pointer
    
    def get_data(self):
        return self.data
    
    def set_left(self, left):
        self.left_pointer = left
    
    def set_right(self, right):
        self.right_pointer = right
    
    def set_data(self, data):
        self.data = data

    def insert(self,data):
        # Insert a value into the BST in the correct position.
        if self.get_data() == None:
            # Empty root: place the value here.
            self.set_data(data)
        elif data < self.get_data():
            # Smaller values go to the left subtree.
            if self.get_left() == None:
                self.set_left(Tree(data)) 
            else:
            # Greater or equal values go to the right subtree.
                self.get_left().insert(data)
        else:
            # Greater or equal values go to the right subtree.
            if self.get_right() == None:
                self.set_right(Tree(data)) 
            else:
            # Greater or equal values go to the right subtree.
                self.get_right().insert(data)

    def in_order_traversal(self):
        # Left, root, right gives values in sorted order for a BST.
        """Recursively perform an in-order traversal of the tree and print data."""
        elements = []
        if self.get_left():     
            elements += self.get_left().in_order_traversal()
        elements.append(self.data)
        if self.get_right():
            elements += self.get_right().in_order_traversal()
        return elements

    def post_order_traversal(self):
        # Left, right, root traversal.
        """Recursively perform an in-order traversal of the tree and print data."""
        elements = []
        if self.get_left():
            elements += self.get_left().post_order_traversal()
        if self.get_right():
            elements += self.get_right().post_order_traversal()
        elements.append(self.data)
        return elements
    
t = Tree()

# Generate sample values to test insertion and traversal.
nodes = [random.randint(0,999) for _ in range(10)]

for _ in nodes:
    t.insert(_)

for _ in t.post_order_traversal():
    print(_)

print('break')

for _ in t.in_order_traversal():
    print(_)
````
