````python
# Task 1

def fac(n):
    if n == 0:
        return 1
    else:
        return n*fac(n-1)

print(362880 == fac(9))
````

````python
# Task 2

def gen_ratings():
    import csv

    with open("ratings.csv") as f:
        data = list(csv.reader(f))[1:]

    names = set([i[0] for i in data])

    ratings = {i:{} for i in names}

    for i in data:
            ratings[i[0]][i[1]] = int(i[-1])

    return ratings
````

## Task 3

To determine how good a seating arrangement is, the total happiness can be used. When calculating the happiness for a student, the students adjacent to them have to be considered. This not only includes the students on the **right** and **left** of the student but also those at the **front** and the **back** of the student. The total happiness is the sum of all the individual happiness levels of all students in the seating arrangement.

Implement a function <span style='color:#3e66f4;'>cal_hap</span><span style='color:#d83439'>(</span>seats,n,ratings<span style='color:#d83439'>)</span> where <span style='color:#ED7117;'>seats</span> is a 1D array of the grid, <span style='color:#ED7117;'>n</span> is the width of the grid and <span style='color:#ED7117;'>ratings</span> is the dictionary object returned by <span style='color:#3e66f4;'>gen_ratings</span><span style='color:#d83439'>(</span><span style='color:#d83439'>)</span>. The function should return an integer for the total happiness of the seating plan.


An example of <span style='color:#ED7117;'>seats</span> will be as such for the arrangement shown above:

```python
["A","B","C","D","F","G","H","I"] . 
```

````python
# Task 3

# Approach 1
def cal_hap(seats,n,ratings):
    tot_hap = 0

    grid = [seats[i:i+n] for i in range(0,n*n,n)]

    transposed = list(zip(*grid))

    combi = [
         grid,
         transposed,
         [list(reversed(i)) for i in grid],
         [list(reversed(i)) for i in transposed]
    ]

    for grid_var in combi:

        for i in range(n):

            for j in range(n-1):

                pov = grid_var[i][j]
                ref = grid_var[i][j+1]

                if pov is not None:
                    pov_ratings = ratings[pov]
                    tot_hap += pov_ratings.get(ref,0)

    return tot_hap


# Approach 2
def cal_hap(seats, n, ratings):
    tot_hap = 0

    def calculate(i, delta):
        new_i = i + delta

        if not (0 <= new_i < len(seats)):
            return 0

        if delta in [-1, 1] and i // n != new_i // n:
            return 0

        their = seats[i]
        them = seats[new_i]

        if their is not None:
            return ratings[their].get(them, 0)

        return 0

    for student in range(len(seats)):
        for delta in [-n, -1, 1, n]:
            tot_hap += calculate(student, delta)

    return tot_hap
````

````python
# Task 6

def happiest_seating(first,ratings):

    n = int(len(ratings.keys())**0.5)

    seats = [None]*(n*n)

    seats[0] = first

    def _helperHap(index,seats,n,ratings,best_hap=0,best_seats = None):

        if index == n*n:
            happiness = cal_hap(seats,n,ratings)

            if happiness > best_hap:
                best_hap = happiness
                best_seats = seats[:]

                return best_hap,best_seats
                
        for person in sorted(list(ratings.keys())):
            if person not in seats:
                seats[index] = person
                best_hap,best_seats = _helperHap(index+1,seats,n,ratings,best_hap,best_seats)
                seats[index] = None


        return best_hap,best_seats


    
    best_hap,best_seats = _helperHap(1,seats,n,ratings)

    grid = [best_seats[i:i+n] for i in range(0,n*n,n)]

    return grid, best_hap

ratings = gen_ratings()
possibleSeats = happiest_seating("A",ratings)

print(possibleSeats)
````

````python
class cQueue:
    def __init__(self,size):
        self.array = [None]*size
        self.tail = -1
        self.head = 0
        self.max_size = size

    def is_empty(self):
        return self.tail +1 == self.head

    def is_full(self):
        return (self.tail + 1)%self.max_size == self.head

    def enqueue(self,data):
        if self.is_full():
            return False
        
        self.tail = (self.tail+1)%self.max_size
        self.array[self.tail] = data
        return True

    def dequeue(self):
        if self.is_empty():
            return False
        dq = self.array[self.head]
        self.head = (self.head+1)%self.max_size
        return dq

    def __str__(self):
        return str(self.array)

cq = cQueue(5)
print(cq)
print(cq.is_full())
print(cq.is_empty())
cq.enqueue(1)
cq.enqueue(2)
cq.enqueue(3)
cq.enqueue(4)
print(cq.head,cq.tail)
cq.enqueue(5)

cq.dequeue()
print(cq)
cq.enqueue(9)
print(cq.head,cq.tail)
print(cq)
print(cq.is_full())
````
