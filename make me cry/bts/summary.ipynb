**CSV format for Audio Sample**

datetime, ampl(un-normalised)



**Process**

1. **Normalisation:** Define the amplitude range: (-1) to (+1)
    * Normalisation gives a known finite range that every sample must fall within. Your normalised audio has:

<center>

$$
-1 \leq A \leq 1
$$

</center>


2. **Quantisation:** Determine the number of quantisation levels
    * Quantisation requires dividing the possible amplitude range into a fixed number of discrete levels. Without a defined range, you cannot determine where a particular amplitude falls. 

<div style="margin-left: 40px;">

For (n) bits:

<center>

$$
L=2^n
$$

</center>
For 8 bits:
<center>

$$
L=2^8=256
$$

</center>

So there are **256 possible amplitude values**.


Map to the current amplitude range([-1,+1]) to ([0,255])

* Your amplitude is currently:$-1\leq A\leq1$

* Binary encoding requires a non-negative integer from:$0\leq Q\leq255$

Linear Mapping:

1. Translate the amplitude rightwards: $Q = A+1$
    * $0\leq Q\leq2$
2. Divide the total range: $Q = \frac{A+1}{2}$
    * $0\leq Q \leq1$
3. Scale by 255: $Q=\frac{A+1}{2}\times255$
    * $0\leq Q\leq255$

To summarise, a **linear mapping** is applied to the normalised data to map it from one numerical range to another:

$$
[-1,1]\rightarrow[0,255]
$$

For example:

```text
-1.0 → 0
 0.0 → 127.5
+1.0 → 255
```


</div>


4. Round to the nearest integer

You will generally get a decimal value.

For example:

$$
A=0.25
$$

gives:

$$
Q=\frac{1.25}{2}\times255
=159.375
$$

You cannot have **159.375 of a quantisation level**, so round:

$$
Q=159
$$

**Why?**
Digital systems have **discrete** values. Quantisation converts the continuous range of possible amplitudes into a finite set of discrete levels.

This introduces **quantisation error**:

$$
\text{error}=A-\text{quantised amplitude}
$$

That error is an unavoidable consequence of representing a continuous signal using a finite number of levels.

---

### 5. Convert the quantisation level to binary

You now have an integer between 0 and 255.

For example:

$$
159_{10}=10011111_2
$$

**Why?**
The quantised value must ultimately be represented using the binary states available to a digital computer.

Because you chose 8-bit quantisation, **every sample must use exactly 8 bits**.

For example:

```text
5 → 00000101
159 → 10011111
```

The leading zeroes are important because every sample occupies the same 8-bit space.

---

### 6. Concatenate the samples

Suppose your first three samples produce:

```text
10011111
01101010
11100001
```

The stored binary data becomes:

```text
100111110110101011100001
```

**Why?**
Each sample contributes 8 bits, so for (N) samples:

$$
\boxed{\text{total bits}=N\times8}
$$

For 7,000 samples:

[
7000\times8=56,000\text{ bits}
]

or:

[
\frac{56,000}{8}=7,000\text{ bytes}
]

So your entire audio sample can be represented by **7,000 bytes** when using 8-bit quantisation.

---

### The complete mathematical process

[
\boxed{
A
\rightarrow
\frac{A+1}{2}\times255
\rightarrow
\operatorname{round}(Q)
\rightarrow
\text{8-bit binary}
}
]

Conceptually:

**Analogue amplitude**

→ **normalised numerical amplitude**

→ **quantised integer**

→ **binary representation**

The important conceptual point for your question is that **you aren't directly converting the floating-point amplitude into binary**. You first **quantise** it into one of (2^n) discrete levels, and *then* represent that level in binary.
