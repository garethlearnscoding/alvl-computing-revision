```python
# Task 1

def factor_sum(x):
    cache = {1}
    for i in range(1,(x**0.5)+1):
        if x%i == 0:
            # print(i)
            # print(cache)
            cache |= {i} | {(i*j) for j in cache if i*j < x and not x%(i*j)}

            
            # print(cache)

    return sum(set(cache))


factor_sum(10000000000000000)
```
