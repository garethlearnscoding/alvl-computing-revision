```python
# Task 1

class Node:
    def __init__(self,value):
        self.value = value
        self.left = None
        self.right = None
```

```python
# Task 2

class Tree:
    def __init__(self,root):
        self.root = root

    def insert(self,node):
        curr = self.root
        while True:
            if node.value < curr.value:
                if curr.left is None:
                    curr.left = node
                    return
                curr = curr.left

            elif node.value > curr.value:
                if curr.right is None:
                    curr.right = node
                    return
                curr = curr.right

    def in_order_traversal(self,root=None,result=None):
        if result is None and root is None:
            result = []
            root = self.root

        if root:
            self.in_order_traversal(root.left,result)
            result.append(root.value)
            self.in_order_traversal(root.right,result)

        return result

    def out_of_place_flatten(self,root=None,LL=None):
        if LL is None and root is None:
            root = self.root
            LL = Tree(Node(root.value))

        if root:
            if root != self.root:
                curr = LL.root
                while curr.right is not None:
                    curr = curr.right
                curr.right = Node(root.value)
            self.out_of_place_flatten(root.left,LL)
            self.out_of_place_flatten(root.right,LL)    

        return LL              


    def flatten(self):
    
        while True: 
            if self.root.left == None: 
                break 
    
            previous = None  
            current = self.root 
    
            while current.left != None: 
                previous = current 
                current = current.left 



            print(f"Previous: {previous.value}")
            print(f"Current: {current.value}")

            root_curr = current
            while current.right != None:
                current = current.right
                    
            print(f"Previous: {previous.value}")
            print(f"Current: {current.value}")

            current.right = previous.right
            previous.right = root_curr
            print(f"Current's right: {current.right.value}")
            print(f"Previous's right: {previous.right.value}")
            previous.left = None

            print()




    def display(self):
        curr = self.root

        while curr is not None:
            yield curr.value
            curr = curr.right   

        

```

```python
# Task 3

BST = Tree(Node(5))
BST.insert(Node(3))
BST.insert(Node(2))
BST.insert(Node(4))
BST.insert(Node(7))
BST.insert(Node(8))

res = BST.in_order_traversal()
print(res)

LL = BST.out_of_place_flatten()
print(*LL.display())

BST.flatten()
print(*BST.display())
```
