```python
# task 3.1

def create_puzzle(n):
    import random
    tiles = [i for i in range(0,n**2)]
    random.shuffle(tiles)

    grid = []

    for _ in range(n):
        row = tiles[:n]
        if 0 in tiles:
            c = tiles.index(0)
            r = _
        grid.append(row)
        tiles = tiles[n::]


    return grid,r,c

for i in range(3,6):
    puzzle,r,c = create_puzzle(i)
    print("Coordinate of space: ({},{})".format(r,c))
    print("Puzzle: " + str(puzzle))
    print()
```

```python
# task 3.2

def display_puzzle(puzzle):
    display = []
    for i in puzzle:
        tmp = []
        for j in i:
            tmp.append(j)
        display.append(tmp)

    n = len(display[0])

    largest_int = n**2-1

    longest_int = len(str(largest_int)) + 3

    display.insert(0,list(range(n)))

    for r in range(n+1):
        display[r].insert(0,r-1)
        for c in range(n+1):
            term = str(display[r][c])
            if term == '0' and c!= 0 and r!= 0:
                term = "#"
            elif term == '-1':
                term = " "
        
            display[r][c] = " "*(longest_int-len(term)) + term

    displayed_grid = list(map(lambda x: "".join(x),display))

    print(*displayed_grid,sep="\n")
        


puzzle = [
    [14,10,9,3],
    [2,4,5,12],
    [13,0,15,8],
    [7,1,6,11]
]
display_puzzle(puzzle)
```

```python
# task 3.3

def find_available_move(puzzle,r,c):
    possible = {
        "U": (r-1,c),
        "D": (r+1,c),
        "L": (r,c-1),
        "R": (r,c+1)

    }

    avail = []

    for m,(r,c) in possible.items():
        try:
            move = puzzle[r][c]
            avail.append(m)
        except:
            pass

    return avail



puzzle = [
    [14,10,9,3],
    [2,4,5,12],
    [13,0,15,8],
    [7,1,6,11]
]

find_available_move(puzzle,3,1)
```

```python
# task 3.4

def make_move(move,puzzle,r,c):
    pr,pc = r,c
    coords = {
        "U": (r-1,c),
        "D": (r+1,c),
        "L": (r,c-1),
        "R": (r,c+1)
    }

    nr,nc = coords[move]

    puzzle[nr][nc],puzzle[pr][pc] = puzzle[pr][pc],puzzle[nr][nc]

    


puzzle = [
    [14,10,9,3],
    [2,4,5,12],
    [13,0,15,8],
    [7,1,6,11]
]

make_move('D',puzzle,2,1)

print(puzzle)
```

```python
# task 3.5

def win(puzzle):
    n = len(puzzle[0])

    solved = []
    start = 0
    for r in range(n):
        tmp = []
        for i in range(start,n+start):
            tmp.append(i+1)
        solved.append(tmp)
        start = n+start

    solved[-1][-1] = "#"

    return puzzle == solved

puzzle = [
    [14,10,9,3],
    [2,4,5,12],
    [13,0,15,8],
    [7,1,6,11]
]

win(puzzle)
```

```python
# task 3.6
from datetime import datetime

menu = "1) New game\n2)Make one move\n3) Make many moves\n4)Quit"

def game_menu():
    valid =["U","D","L","R"]
    puzzle = [
        [2,3,5],
        [1,0,4],
        [7,8,6]
    ]
    r,c = 1,1
    display_puzzle(puzzle)
    print()
    print(menu)
    start = datetime.now()
    while True:
        action = input()
        if action not in ['1','2','3','4']:
            continue

        if action == '1':
            size = None
            while size not in ['3','4','5']:
                size = input("Enter size (3,4,5): ")
            puzzle,r,c = create_puzzle(size)
            start = datetime.now()

        elif action == '4':
            end = datetime.now()
            duration = (start - end).seconds
            print(duration)
            break

        elif action == '2':
            pos = find_available_move(puzzle,r,c)
            move = None
            while move not in ['U','D','L','R']:
                move = input(f"Enter move ({",".join(pos)}): ")
            make_move(move,puzzle,r,c)
            if win(puzzle):
                end = datetime.now()
                duration = (start-end).seconds
                print(f"Won in: {duration}")
                break

        elif action == "3":
            moves = []
            while not all([(m in valid) for m in moves]):
                moves = input("Enter series of moves (U,D,R,L): ")
            for m in moves:
                make_move(m,puzzle,r,c)
                if win(puzzle):
                    end = datetime.now()
                    duration = (start-end).seconds
                    print(f"Won in: {duration}")
                    break


        display_puzzle(puzzle)
        print(menu)



game_menu()


```
