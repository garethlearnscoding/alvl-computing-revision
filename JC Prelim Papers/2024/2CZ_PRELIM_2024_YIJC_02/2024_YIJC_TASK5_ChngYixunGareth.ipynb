```python
# task 5.1

class Node:
    def __init__(self,data):
        self.data = data
        self.ptr = None

    def get_data(self):
        return self.data

    def get_ptr(self):
        return self.ptr

    def set_data(self,val):
        self.data = val

    def set_ptr(self,val):
        self.ptr = val
```

```python
# task 5.2

class Queue:
    def __init__(self):
        self.root = None

    def enqueue(self,new_data):
        if self.root is None:
            self.root = Node(new_data)
        else:
            curr = self.root
            while curr.ptr is not None:
                curr = curr.ptr
            curr.ptr = Node(new_data)

    def dequeue(self):
        if self.root is None:
            print("Queue is empty")
        else:
            item = self.root.data
            self.root = self.root.ptr
            return item

    def peek(self):
        if self.root is None:
            return "Queue is empty"
        else:
            return self.root.data

    def display(self):
        if self.root is None:
            print("Queue is empty")
        else:
            curr = self.root
            while curr is not None:
                print(curr.data)
                curr = curr.ptr
```

```python
# task 5.6

class Queue:
    def __init__(self):
        self.root = None

    def enqueue(self,new_data):
        if self.root is None:
            self.root = Node(new_data)
        else:
            curr = self.root
            while curr.ptr is not None:
                curr = curr.ptr
            curr.ptr = Node(new_data)

    def dequeue(self):
        if self.root is None:
            print("Queue is empty")
        else:
            item = self.root.data
            self.root = self.root.ptr
            return item

    def peek(self):
        if self.root is None:
            return "Queue is empty"
        else:
            return self.root.data

    def display(self):
        if self.root is None:
            print("Queue is empty")
        else:
            curr = self.root
            while curr.ptr is not None:
                print(curr.data)
                curr = curr.ptr
            print(curr.data)

    def update(self,string):
        if self.root is None:
            print("Queue is empty")
        else:
            now = convert(string)
            curr = self.root
            while curr is not None and expirydate(curr.data[1]) < now:
                curr = curr.ptr
            self.root = curr

    def use(self,n):
        if self.root is None:
            print(f'There is insufficient reward points, {n} points not deducted.')
        else:
            while n > 0 and self.root is not None:
                if self.root.data[0] <= n:
                    root = self.dequeue()
                    n -= root[0]
                else:
                    r,d = self.root.data
                    r -= n
                    self.root.data = (r,d)
                    return
            print('There is insufficient reward points, {n} points not deducted.')

RewardQ = Queue()
RewardQ.enqueue(reward1)
RewardQ.enqueue(reward2)
RewardQ.enqueue(reward3)

RewardQ.use(50)
RewardQ.display()
print()
RewardQ.use(75)
RewardQ.display()
print()
RewardQ.use(30)
RewardQ.display()
```
