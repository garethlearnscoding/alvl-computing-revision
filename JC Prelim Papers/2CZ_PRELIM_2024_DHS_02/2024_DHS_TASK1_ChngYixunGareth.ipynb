```python
# task 1.1

def readfile(filename):
    with open(filename) as f:
        server_data = [i.strip() for i in f.readlines()]
    return server_data

server_data = readfile("Audio.txt")
print(server_data)
```

```python
# task 1.2

def compress(bytes_list,bytes_pattern,byte_rep):
    pat_lst = [bytes_pattern[i] + bytes_pattern[i+1] for i in range(0,len(bytes_pattern)-1,2)]
    for i in range(len(bytes_list)):
        if bytes_list[i] == pat_lst[0]:
            subsequent = [bytes_list[i+j] for j in range(len(pat_lst))]
            if subsequent == pat_lst:
                for _ in range(len(pat_lst)):
                    del bytes_list[i]
                bytes_list.insert(i,byte_rep)
                return bytes_list

server_data = readfile("Audio.txt")
compressed = compress(server_data,"636485","c0")
print(compressed)
```

```python
# task 1.3

def decompress(compressed_list,byte_rep,bytes_pattern):
    pat_lst = [bytes_pattern[i] + bytes_pattern[i+1] for i in range(0,len(bytes_pattern)-1,2)][::-1]
    for i in range(len(compressed_list)):
        if compressed_list[i] == byte_rep:
            del compressed_list[i]
            for j in range(len(pat_lst)):
                compressed_list.insert(i,pat_lst[j])
            return compressed_list
        
decompressed = decompress(compressed,"c0","636485")
print(decompressed)
```
