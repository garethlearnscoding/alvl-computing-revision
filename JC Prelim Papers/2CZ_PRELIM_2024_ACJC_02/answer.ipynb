```python
# Task 1.1

def string_value(s1):

    parsed = list(s1.lower())
    vowels = list("aeiou")

    length = len(parsed)
    no_of_vowels = sum([1 for i in parsed if i in vowels])

    return length*no_of_vowels
```

```python
# Task 1.2

def StalinSort(lst):
    for i in range(len(lst)):
        for j in range(len(lst)-1):
            sv_first = string_value(lst[j])
            sv_second = string_value(lst[j+1])
            op = False
            if sv_first == sv_second:
                if lst[j] > lst[j+1]:
                    op = True
            if string_value(lst[j]) > string_value(lst[j+1]):
                op = True

            if op:
                popped = lst.pop(j+1)
                lst.insert(0,popped)

    return lst

print(StalinSort(["Peter","Josh","See","Hi"]))
print(StalinSort(["PetEr","JOsh","Hi","See"]))
print(StalinSort(["Peter","Hello","BYE","Hi","ah","bbbb"]))
```

```python
# Task 1.3

with open("words.txt") as f:
    data = [i.strip() for i in f.readlines()]


print(StalinSort(data))
```

```python
# Task 1.4

def QuickSort(lst):
    if len(lst) <= 1:
        return lst

    import random
    pivot = random.choice(lst)

    if len(set([string_value(i) for i in lst])) == 1:
        lesser = [i for i in lst if i < pivot]
        equal = [i for i in lst if i == pivot]
        greater = [i for i in lst if i > pivot]

        return QuickSort(lesser) + equal + QuickSort(greater)
    else:
        equal = [i for i in lst if string_value(i) == string_value(pivot)]
        lesser = [i for i in lst if string_value(i) < string_value(pivot)]
        greater = [i for i in lst if string_value(i) > string_value(pivot)]

        return QuickSort(lesser) + equal + QuickSort(greater)

print(QuickSort(data))    
```

```python
# Task 1.5

def BinarySearch(lst,value):
    low = 0
    high = len(lst)-1

    while low<high:
        mid = (low+high) // 2

        mid_val = string_value(lst[mid])

        if value == mid_val:
            return mid
        else:
            if value > mid_val:
                low = mid+1
            else:
                high = mid-1
    return False

sorted_data = QuickSort(data)
print(sorted_data)
print(BinarySearch(sorted_data,string_value("lion")))
```

```python
# Task 2.1

def main():
    height = int(input("Enter height of the field: "))
    width = int(input("Enter width of the field: "))

    size_of_field = height*width

    no_of_mines = -1
    while no_of_mines < 0 or no_of_mines > size_of_field:
        no_of_mines = int(input("Enter number of mines: "))

    board = ["X"]*no_of_mines + ["."]*(size_of_field-no_of_mines)
    import random
    random.shuffle(board)

    matrix = [board[i*width:(i*width)+width] for i in range(height)]

    return height,width,matrix

    
main()
```

```python
# Task 2.2

def display(matrix):
    print(*matrix,sep="\n")

def move():
    x = int(input("Enter row: "))
    y = int(input("Enter column: "))

    return y,x

def play():
    height,width,matrix = main()

    display(matrix)

    y,x = move()

    if matrix[y][x] == "X":
        print("Game Over!")
        display(matrix)
        return
```
