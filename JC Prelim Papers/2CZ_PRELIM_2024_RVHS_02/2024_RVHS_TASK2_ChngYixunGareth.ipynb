```python
# task 2.1

def task_2_1(filename):
    with open(filename) as f:
        data = f.read().replace(" ","").strip().split(",")

    characters = [chr(int(d)) for d in data]
    return characters

print(task_2_1("inputA.txt"))
```

```python
# task 2.2

char_lst = task_2_1("inputA.txt")

def task_2_2(char_lst):
    freq = {}
    for c in char_lst:
        if c not in freq.keys():
            freq[c] = 1
        else:
            freq[c] += 1

    data = list(freq.items())

    for i in range(len(data)):
        while i > 0 and data[i][1] > data[i-1][1]:
            data[i],data[i-1] = data[i-1],data[i]
            i -= 1
    
    print("Char  Freq")
    for c,f in data:
        print(f"{f"'{c}'":>4}{f:>6}")



task_2_2(char_lst)
```

```python
# task 2.3

def task_2_3(char_lst):
    n = int(len(char_lst)**0.5)

    flip_dict = {
        "v":"h",
        "h":"v"
    }

    total_moves = n*2 -1
    v= 0
    h = 0
    type = "v"

    r_c_count = []
    f=0
    for i in range(n):
        if i%2 == 0:
            r_c_count.append(f)
        else:
            r_c_count.append((n-1)-f)
            f+=1
    
    matrix = [[None]*n for _ in range(n)]

    for i in range(total_moves):
        no_ele = n - h if type == "v" else n-v

        eles = char_lst[0:no_ele]
        char_lst = char_lst[no_ele:]        

        fixed_axis = r_c_count[v] if type == "v" else n-1-r_c_count[h]

        # print("fa",fixed_axis)

        coords = [(i,fixed_axis) if type =="v" else (fixed_axis,i) for i in range(n)]

        # print(coords)
        if type=="v":
            if v%2 != 0:
                coords = coords[::-1]
                # print("v odd")
            else:
                # print("v even")
                ...
            top = h//2
            bot = h-top
            # print(top,bot)
            del coords[:bot]
            coords = coords[::-1]
            del coords[:top]
            coords = coords[::-1]
            
        else:
            if h%2 != 0:
                coords = coords[::-1]
                # print("h odd")
            else:
                # print("h even")
                ...
            top = v//2
            bot = v-top
            # print(top,bot)
            del coords[:bot]
            coords = coords[::-1]
            del coords[:top]
            coords = coords[::-1]

        # print(eles,coords)

        for i in range(no_ele):
            r,c = coords[i]
            ele = eles[i]
            matrix[r][c] = ele

        # print()
        # print(*matrix,sep="\n")
        # print()

        if type == "v":
            v += 1
        else:
            h+=1 
        type = flip_dict[type]

    return matrix


char_lst = task_2_1("inputA.txt")

matrix = task_2_3(char_lst)

result = task_2_3(task_2_1("example.txt"))
for row in result:
    for char in row:
        print(char,end="")
    print()



```

```python
# task 2.4
import csv

def task_2_4(input_file, outputfile):
    char_lst = task_2_1(input_file)

    matrix = task_2_3(char_lst)

    with open(outputfile,"w",newline="") as f:
        writer = csv.writer(f)
        writer.writerows(matrix)

task_2_4("example.txt","output.txt")
task_2_4("inputA.txt","outputA.txt")
task_2_4("inputB.txt","outputB.txt")

```
