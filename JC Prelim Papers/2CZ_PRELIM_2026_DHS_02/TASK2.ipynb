```python
# Task 2.1

class Node:
    def __init__(self):
        self.data = ""
        self.pointer = -1

    def set_data(self,data):
        self.data = data

    def get_data(self):
        return self.data

    def set_pointer(self,pointer):
        self.pointer = pointer

    def get_pointer(self):
        return self.pointer
```

```python
# Task 2.2

class SFSS:
    def __init__(self,n):
        array = [Node() for _ in range(n)]
        for i in range(n-1):
            array[i].set_pointer(i+1)

        self.array = array

        self.directory = {}

        self.free = 0

    def set_free(self,free):
        self.free = free

    def get_free(self):
        return self.free

    def get_array(self):
        return self.get_array

    def get_directory(self):
        return self.directory

    def save_file(self,filename,datastring):
        if self.free == -1:
            print("No available space to save file.")
            return
        
        curr = None
        for char in datastring:
            curr = self.free if curr is None else self.array[curr].get_pointer()
            self.array[curr].set_data(char)

        self.directory[filename] = self.free

        self.free = self.array[curr].get_pointer()
        self.array[curr].set_pointer(-1)


    def delete_file(self,filename):
        curr = self.directory.get(filename)   

        if curr is None:
            print("File not found")
            return

        while curr != -1:
            tmp = self.array[curr].get_pointer()
            self.array[curr].set_pointer(self.free)
            self.array[curr].set_data("")
            self.free = curr
            curr = tmp

        del self.directory[filename]

    def display(self):
        print(f"File Directory: {self.directory}")
        print(f"Free space pointer: {self.free}")
        print(f"Array State:")
        print("Index | Data | Next")
        for idx,node in enumerate(self.array):
            print(f"{idx:>5} | {node.get_data():<4} | {node.get_pointer():<4}")
        
```

```python
# Task 2.3

files = SFSS(10)
files.save_file("file1","ace")
files.save_file("file2","man")
files.delete_file("file1")
files.save_file("file3","nudhigh")
files.display()
```
