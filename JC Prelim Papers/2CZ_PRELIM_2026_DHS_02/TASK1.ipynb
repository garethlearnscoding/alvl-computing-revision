```python
# Task 1.1

def load_key(filename):
    with open(filename) as f:
        key = f.read().split(",")

    int_key = [int(i) for i in key]

    return tuple(int_key)

print(load_key("publickey.txt"))
print(load_key("privatekey.txt"))
```

```python
# Task 1.2

def load_message(filename):
    with open(filename) as f:
        msg = f.read()
    return msg

print(load_message("plaintext.txt"))
```

```python
# Task 1.3

def encrypt_message(msg,e,n):
    encrypted_values = []
    for c in msg:
        encrypted_int = (ord(c)**e)%n
        encrypted_values.append(encrypted_int)
    return encrypted_values
```

```python
# Task 1.4

def decrypt_message(encrypted_values,d,n):
    decrypted_msg = ""
    for value in encrypted_values:
        decrypted_character = chr((value**d)%n)
        decrypted_msg += decrypted_character
    return decrypted_msg
```

```python
# Task 1.5

private_key,private_mod = load_key("privatekey.txt")
public_key,public_mod = load_key("publickey.txt")

msg = load_message("plaintext.txt")

encrypted_values = encrypt_message(msg,private_key,private_mod)

decrypted_msg = decrypt_message(encrypted_values,public_key,public_mod)

print(decrypted_msg)
```
