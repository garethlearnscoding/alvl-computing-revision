```python
# task 4.1

tables = [
    """
CREATE TABLE IF NOT EXISTS Flight (
    FlightNo TEXT PRIMARY KEY,
    DepartCity TEXT,
    ArrivalCity TEXT,
    DepartTime TEXT,
    ArrivalTime TEXT
)
    """,
    """
CREATE TABLE IF NOT EXISTS Customer (
    CustomerNo INTEGER PRIMARY KEY,
    Name TEXT,
    Gender TEXT,
    DOB TEXT
)
    """,
    """
CREATE TABLE IF NOT EXISTS Ticket (
    TicketNo INTEGER PRIMARY KEY,
    CustomerNo INTEGER REFERENCES Customer(CustomerNo) ON DELETE CASCADE,
    FlightNo TEXT REFERENCES Flight(FlightNo) ON DELETE CASCADE,
    Date TEXT,
    Seat TEXT
)
    """
]

import sqlite3
con = sqlite3.connect("TRIP.db")
c = con.cursor()

c.executescript(";".join(tables))

con.commit()
con.close()
```

```python
# task 4.2
import csv

with open("FLIGHT.txt") as f:
    flight_data = list(csv.reader(f))

with open("CUSTOMER.txt") as f:
    customer_data = list(csv.reader(f))

with open("TICKET.txt") as f:
    ticket_data = list(csv.reader(f))

flight_query = "INSERT INTO Flight VALUES(?,?,?,?,?)"
customer_query = "INSERT INTO Customer VALUES (?,?,?,?)"
ticket_query = "INSERT INTO Ticket VALUES (?,?,?,?,?)"

import sqlite3
con =sqlite3.connect("TRIP.db")
c = con.cursor()

c.executescript("""
DELETE FROM Flight;
DELETE FROM Customer;
DELETE FROM Ticket
""")

c.executemany(flight_query,flight_data)
c.executemany(customer_query,customer_data)
c.executemany(ticket_query,ticket_data)

con.commit()
con.close()
```
