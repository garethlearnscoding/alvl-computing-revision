```python
# task 1

def task1_1(key,message):
    if key == "" or message == "":
        return ""
    message = message.replace(" ","z") + "x"*(len(key) - len(message)%len(key))
    columns = [[k,""] for k in key]
    for c,m in enumerate(message):
        idx = c%len(key)
        columns[idx][1] += m
    columns.sort(key=lambda x: x[0])
    return "".join([i[1] for i in columns])
```

```python
# task 1

def soln(key,message):
    if "" in [key,message]:
        return False
    message = message.replace(" ","z") + "x"*(-len(message)%len(key))
    strings = [(key[c%len(key)],m) for c,m in enumerate(message)]
    strings = sorted(strings,key=lambda x: x[0])
    return "".join([i[1] for i in strings])
```

```python
cipher = soln("","")
print(cipher)
```
