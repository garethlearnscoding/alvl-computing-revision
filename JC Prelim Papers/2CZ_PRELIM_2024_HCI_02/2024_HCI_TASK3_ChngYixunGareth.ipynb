```python
# task 3.3

class PriorityQueue:
    def __init__(self):
        self.root = None

    def isEmpty(self):
        return self.root == None

    def viewPendingTasks(self):
        if self.isEmpty():
            print("Priority Queue is empty.")
        else:
            curr = self.root
            while curr is not None:
                print(curr)
                curr = curr.next

    def enqueue(self,taskNode):
        if self.isEmpty() or taskNode.priority > self.root.priority:
            temp = self.root
            self.root = taskNode
            taskNode.next = temp
        else:
            prev = self.root
            curr = self.root.next
            while curr is not None and taskNode.priority < curr.priority:
                prev = curr
                curr = curr.next
            prev.next = taskNode
            taskNode.next = curr

    def dequeue(self):
        if self.isEmpty():
            print("Priority Queue is empty.")
        else:
            item = self.root
            self.root = self.root.next
            return item

    def delete(self,taskName):
        if self.isEmpty():
            print("Priority Queue is empty.")
        elif self.root.taskName == taskName:
            self.root = self.root.next
        else:
            prev = self.root
            curr = self.root.next
            while curr is not None:
                if curr.taskName == taskName:
                    prev.next = curr.next
                    return True
                prev = curr
                curr = curr.next
            return False
        
    def search(self,taskName):
        if self.isEmpty():
            print("Priority Queue is empty.")
        elif self.root.taskName == taskName:
            return self.root
        else:
            curr = self.root.next
            while curr is not None:
                if curr.taskName == taskName:
                    return curr
                curr = curr.next
            print("Task not Found")

    def updatePriority(self,taskName,newPriority):
        if self.delete(taskName):
            self.enqueue(TaskNode(taskName,newPriority))
            print(f"Priority of {taskName} has been updated to {newPriority}.")
            return True
        return False
```

```python
# task 3.3

pQ = PriorityQueue()

with open("TASKS.txt") as f:
    data = [[i[0],int(i[1])] for i in csv.reader(f)]

for i in data:
    pQ.enqueue(TaskNode(*i))

pQ.dequeue()
pQ.dequeue()

result = pQ.search('Task 1')
print("Search Result:",result)
print()
result = pQ.delete('Task 5')
print("Delete Result:",result)
print()
result = pQ.updatePriority('Task 4',4)
print("Update Result:",result)
```
