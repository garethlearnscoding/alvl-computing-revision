```python
# task 2.1

def read_csv(filename):
    import csv

    def parse(line):
        red,green,blue,name = line
        return [int(red),int(green),int(blue),name]
    
    with open(filename) as f:
        data = list(map(parse,[i for i in csv.reader(f)]))

    return data

colours = read_csv("paint.csv")
print(len(colours))
print(colours)
```

```python
# task 2.2

def find(array,r,g,b):
    for colour in array:
        if [r,g,b] == colour[:-1]:
            return colour[-1]
    return 'nil'

print(find(colours,253,253,5))
print(find(colours,5,253,253))
```

```python
# task 2.3

def rgb(array,colour):
    idx_dict = {'red':0,'green':1,'blue':2}
    if colour not in idx_dict.keys():
        print('Invalid Colour Provided')
        return []
    else:
        idx = idx_dict[colour]
        for i in range(len(array)):
            for j in range(len(array)-i-1):
                if array[j][idx] < array[j+1][idx]:
                    array[j],array[j+1] = array[j+1],array[j]
        return array
            
print(rgb(colours,'yellow'))
print(rgb(colours,'blue'))
```

```python
# task 2.4

def bright(array):

    def brightness(colour):
        r,g,b,_ = colour
        return r+g+b
    
    if len(array) <= 1:
        return array

    import random
    pivot = brightness(random.choice(array))
    lesser = [i for i in array if brightness(i) < pivot]
    greater = [i for i in array if brightness(i) > pivot]
    equal = [i for i in array if brightness(i) == pivot]

    return bright(greater) + equal + bright(lesser)

print(bright(colours))
```

```python
# task 2.5

def menu():
    colours = read_csv('paint.csv')
    pref = None
    while pref not in ["red","green","blue","dark","bright"]:
        pref = input("Enter your preference (red,green,blue,dark or bright):")
    if pref in ['red','green','blue']:
        return rgb(colours,pref)[:3]
    elif pref == "bright":
        return bright(colours)[:3]
    elif pref == "dark":
        return bright(colours)[::-1][:3:]


print(menu())
print(menu())
print(menu())
```
