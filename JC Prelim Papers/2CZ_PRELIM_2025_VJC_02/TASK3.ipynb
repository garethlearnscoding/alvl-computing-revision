```python
# Task 1

class Node:
    def __init__(self,team):
        self.team = team
        self.left_ptr = None
        self.right_ptr = None

    def __str__(self):
        return self.team

    def display(self):
        return self.team
```

```python
# Task 2
import math

class TournamentTree:
    def __init__(self):
        self.root = None

    def create_tree(self,num_of_teams):
        lst = [Node(f"Team {i}") for i in range(num_of_teams)]
    
        match_no = 1
        layer = 1

        while layer<=math.log(num_of_teams)+1:

            print(f"Layer: {layer}")
            no_of_iter = num_of_teams//(2**layer)

            print([i.display() for i in lst])
            print()

            for _ in range(0,no_of_iter):
                print([i.display() for i in lst])
                tmp = Node(f"Match {match_no}")

                tmp.left_ptr = lst.pop(0)
                tmp.right_ptr = lst.pop(0)

                lst.append(tmp)

                match_no += 1

            layer += 1

        self.root = lst[0]

    def create_tree(self,num_of_teams):
        lst = [Node(f"Team {i}") for i in range(num_of_teams)]
    
        match_no = 1
        layer = 1

        while layer<=math.log(num_of_teams)+1:

            no_of_iter = num_of_teams//(2**layer)

            for _ in range(0,no_of_iter):

                tmp = Node(f"Match {match_no}")

                tmp.left_ptr = lst.pop(0)
                tmp.right_ptr = lst.pop(0)

                lst.append(tmp)

                match_no += 1

            layer += 1

        self.root = lst[0]


    def level_order(self):
        '''
        print tree from root
        print out all the nodes of the same level in a row 
        '''
        self._level_order(self.root)
        
    def _level_order(self, node_list):
        '''
        Helper function for internal use
        print tree from node list
        '''
        # convert node_list to a list if it is not
        if not isinstance(node_list, list):
            node_list = [node_list]
        # Stop recursion if the list is empty
        if not node_list:
            return
        # define a list to collect node in next layer
        next_layer = []
        while node_list:
            node = node_list.pop(0) # Remove the first item
            print(node, end = ' ')
            if node.left_ptr:
                next_layer.append(node.left_ptr)
            if node.right_ptr:
                next_layer.append(node.right_ptr)
        print() # print new line after processing a layer
        self._level_order(next_layer)
        

TT = TournamentTree()
TT.create_tree(8)  
TT.level_order()             

```
