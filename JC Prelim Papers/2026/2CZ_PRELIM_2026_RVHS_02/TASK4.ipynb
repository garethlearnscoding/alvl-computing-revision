```python
# Task 4.2
import sqlite3

with sqlite3.connect("got_talent.db") as con:
    c = con.cursor()

    c.execute("""
    SELECT pt.id, pt.description, (max_number - COUNT(part.performance)) 
    FROM performance_types pt
    LEFT OUTER JOIN participants part ON pt.id = part.performance
    GROUP BY pt.id
    """)

    data = list(c.fetchall())

print(data)
```
