```python
import random

def createSensorNode():
    sensorID = random.randint(0,5)
    sensorID = f"{sensorID}"

    return {"Left":None, "SensorID":sensorID, "Right":None}
```

```python
def insert(Root,newSensorNode,success=True,ogRoot = None): 

    if Root == None: 
        Root = newSensorNode 
        return Root,success
    
    if ogRoot is None:
        ogRoot = Root

    if success == False:
        newSensorNode = createSensorNode()
        

    if Root['SensorID'] > newSensorNode['SensorID']: 
        Root["Left"],success = insert(Root["Left"], newSensorNode,success,ogRoot)

    elif Root['SensorID'] < newSensorNode['SensorID']: 
        Root["Right"],success= insert(Root["Right"],newSensorNode,success,ogRoot)

    else:
        success = False
        return None,success

        
    return Root,success



Troot = None

from pprint import pprint

for i in range(5):
    Node = createSensorNode()
    Troot,_ =insert(Troot,Node)
    pprint(Troot)
    print()
```
