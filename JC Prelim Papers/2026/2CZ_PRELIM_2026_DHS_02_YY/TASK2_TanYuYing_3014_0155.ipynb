```python
#Task 2.1 

class Node: 
    def __init__(self): 
        self.data = "" 
        self.pointer = -1 

    def set_data(self,data): 
        self.data = data 
    def set_pointer(self,pointer): 
        self.pointer = pointer 

    def get_data(self): 
        return self.data 
    def get_pointer(self): 
        return self.pointer 
```

```python
#Task 2.2 

class SFSS: 
    def __init__(self,n): 
        self.array = [Node() for _ in range(n)] 
        self.directory = {} 
        self.free = 0 

        for i in range(len(self.array)-1): 
            self.array[i].pointer = i + 1 

    def set_free(self,free): 
        self.free = free 
    def get_free(self): 
        return self.free 

    def get_array(self): 
        return self.array 

    def get_directory(self): 
        return self.directory 

    def save_file(self,filename,datastring): 
        if self.free == -1: 
            print("Linked list is full") 
            return 

        start_node = self.free 

        for i in range(len(datastring)): 
            temp = self.free 

            self.array[temp].set_data(datastring[i]) 

            self.free = self.array[temp].pointer 

            if i == len(datastring) - 1: 
                self.array[temp].pointer = -1 

        self.directory[filename]=start_node 

    def delete_file(self,filename): 
        if filename not in self.directory.keys(): 
            print(f"{filename} does not exists") 
            return 

        curr = self.directory[filename] 

        while curr != -1: 
            temp = curr 

            self.array[temp].set_data("") 

            curr = self.array[temp].pointer 

            self.array[temp].pointer = self.free 

            self.free = temp 

        self.directory.pop(filename) 

    def display(self): 
        print(f"File Directory: {self.directory}") 
        print(f"Free space pointer: {self.free}") 

        print("Array State: ") 
        print(f"{"Index":^7}|{"Data":^6}|{"Next":^6}") 

        for i in range(len(self.array)): 
            print(f"{(str(i)+" "):>7}|{(" " + self.array[i].data):<6}|{(" " + str(self.array[i].pointer)):<6}")  
```

```python
#Task 2.3 

file_storage = SFSS(10) 

file_storage.save_file("file1","ace") 

file_storage.save_file("file2","man") 

file_storage.delete_file("file1") 

file_storage.save_file("file3","nudhigh")

file_storage.display() 
```
