```python
#Task 1.1 

def load_key(filename): 
    with open(filename) as f: 
        key = tuple(int(x.strip()) for x in f.read().split(" , ")) 

    return key
```

```python
#Task 1.2 

def load_message(filename):  
    with open(f"{filename}") as f: 
        message = f.read()

    return message 
```

```python
#Task 1.3 

def encrypt_message(message,e,n): 
    encrypt_int_lst = [] 

    for char in message: 
        val = (ord(char)**e) % n  
        encrypt_int.append(val) 

    return encrypt_int 
        
```

```python
#Task 1.4 

def decrypt_message(lst,d,n): 
    decrypted_message = "" 

    for val in lst: 
        decrypted_char = chr((val**d) % n) 
        decrypted_message += decrypted_char 

    return decrypted_message 
```

```python
#Task 1.5 

public_key = load_key("publickey.txt") 

private_key = load_key("privatekey.txt") 

print(f"Public Key: {public_key}, Private Key: {private_key}") 

message = load_message("plaintext.txt") 

print(f"Plaintext message: {message}") 

encrypt_lst = encrypt_message(message,public_key[0],public_key[1]) 

print(encrypt_lst) 

decrypted_message = decrypt_message(encrypt_lst,private_key[0], private_key[1]) 

print(f"Decrypted Message: {decrypted_message}") 
```
