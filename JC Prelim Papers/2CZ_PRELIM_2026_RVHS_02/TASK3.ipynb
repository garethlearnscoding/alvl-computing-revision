```python
def task3_1():

    # Initially exists as a 1D list
    letter_lst = []
    invalid = False

    while len(letter_lst) != 25:
        msg = "Key in 5 unique letters that you have not entered before: " if not invalid else "Try again: "
        letters = input(msg)

        if len(letters) != 5: 
            print("You must type 5 letters each time.")
            invalid = True

        if letters.isalpha() == False: 
            print("You must only input letters.")
            invalid = True

        if len(set(letters)) != 5: 
            print("You must only input unique letters.")
            invalid = True

        repeated_letters = [i for i in letters if i in letter_lst]

        if len(repeated_letters) != 0:
            print(f"You have keyed in {"".join(set(repeated_letters))} before.")
            invalid = True

        if not invalid:
            letter_lst += list(letters)
            invalid = False

    output = [letter_lst[i*5:(i*5)+5] for i in range(5)]

    return output


task3_1()
```
