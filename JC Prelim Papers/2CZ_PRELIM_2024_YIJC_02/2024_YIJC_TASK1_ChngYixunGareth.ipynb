```python
# task 1.1

def hash(address):
    temp = 9619
    for c in address:
        temp *= 37
        temp += ord(c)

    return temp % 250

hash('ABC')
```

```python
# task 1.2
hash_table = [None]*250

with open("ADDRESSES.TXT") as f:
    data = [i.strip() for i in f.readlines()]

uninserted = []

for i in data:
    idx = hash(i)
    if hash_table[idx] is not None:
        uninserted.append(i)
    else:
        hash_table[idx] = (idx,i)

print(*hash_table[:10],sep="\n")

print("Number of Address in Uninserted List:",len(uninserted))
```

```python
# task 1.3
hash_table = [None]*250

with open("ADDRESSES.TXT") as f:
    data = [i.strip() for i in f.readlines()]

uninserted = []

for i in data:
    hashed = hash(i)
    idx = hashed
    while hash_table[idx] is not None:
        idx = (idx + 1)%250
    hash_table.append((hashed,i))

print(*hash_table[-10:],sep="\n")
```
