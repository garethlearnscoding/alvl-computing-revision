```python
# task 3.1

with open("GRID1.TXT") as f:
    grid1 = [i.strip().split(",") for i in f.readlines()]
```

```python
# task 3.2

def print_it(array):
    for r in array:
        for c in r:
            if c == '':
                c = "."
            print(c,end=" ")
        print()

print_it(grid1)
```

```python
# task 3.3

def user_input():
    action = None
    while action not in [str(i) for i in range(1,8)]:
        action = input("Enter a column (1 - 7): ")
    return int(action)
```

```python
# task 3.4

def insertO(array,col):
    inserted = False
    for i in range(len(array)-1,0-1,-1):
        if array[i][col-1] ==  '' and not inserted:
            array[i][col-1] = "O"
            inserted = True
    if not inserted:
        print("Player (O) missed a turn.")
    return array   
```

```python
# task 3.5

def insertX(array,col):
    inserted = False
    # print(len(array))
    for i in range(len(array)-1,-1,-1):
        if array[i][col-1] ==  '' and not inserted:
            array[i][col-1] = "X"
            inserted = True
    if not inserted:
        print("Player (X) missed a turn.")
    return array   
```

```python
# task 3.6

def checkrow(array):
    for r in array:
        s = "".join(r)
        if "OOOO" in s or "XXXX" in s:
            return True
    return False
```

```python
# task 3.7

def checkcolumn(array):
    transposed = list(zip(*array))
    return checkrow(transposed)


def checkdiagonal(array):
    for row in range(3):
        for col in range(4):
            
            diagonal1 = []
            diagonal2 = []
            for i in range(4):
                diagonal1.append(array[row+i][col+i])
                diagonal2.append(array[row+3-i][col+i])

            string = ''.join(diagonal1)
            if 'OOOO' in string or 'XXXX' in string:
                return True
            else:
                string = ''.join(diagonal2)
                if 'OOOO' in string or 'XXXX' in string:
                    return True
    return False
```

```python
# task 3.8

with open("GRID2.TXT") as f:
    grid = [i.strip().split(",") for i in f.readlines()]
    
while True:

    print_it(grid)
    user_action = user_input()
    grid = insertO(grid,user_action)
    print_it(grid)
    if checkrow(grid) or checkcolumn(grid) or checkdiagonal(grid):
        print("Player Win")
        break

    import random
    com_action = random.randint(1,7)

    grid = insertX(grid,com_action)
    print_it(grid)

    if checkrow(grid) or checkcolumn(grid) or checkdiagonal(grid):
        print("Computer Win")
        break

    print()

```
