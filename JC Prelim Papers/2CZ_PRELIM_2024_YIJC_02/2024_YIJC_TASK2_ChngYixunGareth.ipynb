```python
"""
<A> lo <- lo + 1
<B> hi <- hi -1
<C> temp <- seq[start]
    seq[start] <- seq[hi]
    seq[hi] <- temp

"""

def quicksort(lst):
    quicksort_helper(lst,0,len(lst)-1)
    return lst

def quicksort_helper(lst,start,end):
    if start <end:
        mid = partition(lst,start,end)
        quicksort_helper(lst,start, mid-1)
        quicksort_helper(lst,mid+1,end)
    return lst

def partition(lst,start,end):
    pivotValue = lst[start]
    lo = start + 1
    hi = end
    done = False

    while not done:
        while lo <= hi and lst[lo] <= pivotValue:
            lo += 1
        while lst[hi] >= pivotValue and hi >= lo:
            hi -=1

        if hi < lo:
            done = True
        else:
            lst[lo],lst[hi]=lst[hi],lst[lo]
        
    lst[start],lst[hi]=lst[hi],lst[start]
    return hi
```

```python
# task 2.2

with open("POSTALCODES.TXT") as f:
    data = [int(i.strip()) for i in f.readlines()]

data = quicksort(data)

print(*data[:5],sep='\n')
```
