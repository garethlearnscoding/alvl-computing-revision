```python
def score_guess(secret,guess): 
    if is_valid(secret) and is_valid(guess): 
        hits = 0 
        pseudo_hits = 0 

        hit_letter_index = [] 
        pseudo_letters = [] 

        secret_lst = list(secret)

        for i in range(len(guess)): 
            if guess[i] == secret[i]: 
                hits += 1 
                hit_letter_index.append(i) 


        for i in range(len(guess)): 
            valid_count = True 

            if guess[i] in secret_lst: 
                for index in hit_letter_index: 
                    if guess[i] == secret[index]: 
                        valid_count = False 

                if guess[i] in pseudo_letters: 
                    valid_count = False 

            else: 
                valid_count = False 

            if valid_count: 
                pseudo_hits += 1 
                pseudo_letters.append(guess[i])

    return (hits,pseudo_hits)

print(score_guess("RGBY", "GGRR"))
print(score_guess("RGGB", "RGGB"))
print(score_guess("RGGB", "YYYY"))
```
