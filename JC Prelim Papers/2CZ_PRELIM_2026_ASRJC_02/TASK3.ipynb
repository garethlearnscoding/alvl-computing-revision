```python
class Node:
    def __init__(self,data):
        self.data = data
        self.left_ptr = None
        self.right_ptr = None

    def average(self):
        if len(self.data) != 2:
            return 0
        values = self.data[1]
        average = sum(values)/len(values)
        return round(average,1)

    def display_info(self):
        print(f"Class group : {self.data[0]}\nAverage GP score : {self.data[1]}")
```

```python
from show_recursion_tree import show_recursion_tree

class GPTree:
    def __init__(self):
        self.root = None

    def insert(self,data):
        node = Node(data)

        if self.root is None:
            self.root = node
            return
        
        curr = self.root
        while True:
            if node.average() < curr.average():
                if curr.left_ptr is None:
                    curr.left_ptr = node
                    return
                curr = curr.left_ptr
            else:
                if curr.right_ptr is None:
                    curr.right_ptr = node
                    return
                curr = curr.right_ptr

    def inorder(self):
        result = []
        self._inorder(self.root,result)
        return result
    
    def _inorder(self,node,result):
        if node:
            self._inorder(node.left_ptr,result)
            result.append(node.data)
            self._inorder(node.right_ptr)

    def find_min(self):
        return self._find_min(self.root)

    def _find_min(self,node):
        if node.left:
            self._find_min(node.left_ptr)
        else:
            return node

    def find_max(self):
        return self._find_max(self.root)

    def _find_max(self,node):
        if node.right:
            self._find_max(node.right_ptr)
        else:
            return node

    def depth(self):
        return self._depth(self.root)
    
    @show_recursion_tree
    def _depth(self, node):
        if node is None:
            return 0
    
        return 1 + max(
            self._depth(node.left_ptr),
            self._depth(node.right_ptr)
        )
    
    def depth1(self):
        cache = []
        self._depth(self.root, cache)
        return max(cache)

    def _depth1(self, node, cache, count=0):
        if node:
            count += 1

            if node.left_ptr is None and node.right_ptr is None:
                cache.append(count)
                return

            self._depth(node.left_ptr, cache, count)
            self._depth(node.right_ptr, cache, count)

    def tree_avg(self):
        no_of_students,avg = self._tree_avg(self.root,0,0)
        return avg/no_of_students

    def _tree_avg(self,node,no_of_students,avg): 
        if node is None:
            return [0,0]      
        if node:
            return [sum(i) for i in zip(
                self._tree_avg(node.left_ptr),
                self._tree_avg(node.right_ptr),
                list(node.data)
            )]        
            
```
