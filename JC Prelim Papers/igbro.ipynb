```python
def binarySearch(arr,seek,low=None,high=None):
    if low is None and high is None:
        low = 0
        high = len(arr)

    mid = (low+high)//2

    if low>high:
        return False

    if arr[mid] == seek:
        return mid
    else:
        if seek > arr[mid]:
            return binarySearch(arr,seek,mid+1,high)
        else:
            return binarySearch(arr,seek,low,mid-1)
```
