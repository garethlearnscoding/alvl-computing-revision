```python
# Task 2.1

class Combatant:
    def __init__(self,name,health,damage,cooldown):
        self.name = name
        self.health = int(health)
        self.damage = int(damage)
        self.cooldown = int(cooldown)

    def take_damage(self,points):
        self.health -= points

    def is_dead(self):
        return self.health <= 0
```

```python
# Task 2.2

class Slot:
    def __init__(self,combatant):
        self.combatant = combatant
        self.turns_before_ready = self.combatant.cooldown
        self.next_slot = None

    def get_priority(self):
        return self.turns_before_ready

    def update_priority(self):
        if self.turns_before_ready > 0:
            self.turns_before_ready -= 1
    
```

```python
# Task 2.3

class PriorityQueue:
    def __init__(self):
        self.head = None
        self.count = 0

    def add(self,combatant):
        tmp = Slot(combatant)

        if self.head is None or self.head.turns_before_ready > tmp.turns_before_ready:
            tmp.next_slot = self.head
            self.head = tmp
            self.count += 1
            return
        
        prev = None
        curr = self.head
        while curr is not None and curr.turns_before_ready <= tmp.turns_before_ready:
            prev = curr
            curr = curr.next_slot
        
        prev.next_slot = tmp
        tmp.next_slot = curr

        self.count += 1

    def remove(self,name):
        if self.head is None:
            return
        
        elif self.head.combatant.name == name:
            self.head = self.head.next_slot

        else:
            prev = None
            curr = self.head
            while curr is not None and curr.combatant.name != name:
                prev = curr
                curr = curr.next_slot
            if curr is None:
                return
            else:
                prev.next_slot = curr.next_slot

        self.count -= 1

    def dequeue_highest_priority(self):
        if self.head is None or int(self.head.turns_before_ready) > 0:
            return

        else:
            pop = self.head
            self.head = self.head.next_slot
            self.count -= 1
            return pop.combatant

    def update_slots(self):
        # print("Updating slots...")
        curr = self.head
        while curr is not None:
            curr.update_priority()
            curr = curr.next_slot

    def get_alive_combatants(self):
        alive_combatants = []
        curr = self.head
        while curr is not None:
            alive_combatants.append(curr.combatant)
            curr = curr.next_slot
        return alive_combatants

        


    
```

```python
# Task 2.4

class Player(Combatant):
    def get_opponent_choice(self,opponents):
    
        choice = None
        while choice is None:
            for idx,combatant in enumerate(opponents,start=1):
                print(f"{idx}: {combatant.name}")
            choice = input(f"Choose an opponent: ")
            if choice.isnumeric() and 1 <= int(choice) <= len(opponents):
                return opponents[int(choice)-1]

        return None
        
        

class Enemy(Combatant):
    def get_opponent_choice(self,combatants):
        players = [i for i in combatants if isinstance(i,Player)]
        return players[0]


import csv
with open("COMBATANTS.txt") as f:
    data = list(csv.reader(f))[1:]

   

combatants = []
for entry in data:
    if entry[0] == "player":
        combatants.append(Player(*entry))
    else:
        combatants.append(Enemy(*entry))

PQ = PriorityQueue()
for combatant in combatants:
    PQ.add(combatant)

```

```python
# Task 2.5
remaining_combatants = combatants

def game_won(remaining_combatants):
    return remaining_combatants \
    and all(isinstance(c,Player) for c in remaining_combatants)


while not game_won(remaining_combatants):
    active_combatant = PQ.dequeue_highest_priority()
    if active_combatant is None:
        PQ.update_slots()
    else:
        alive_combatants = PQ.get_alive_combatants()
        # print(alive_combatants)
        
        victim = active_combatant.get_opponent_choice(alive_combatants)
        victim.take_damage(active_combatant.damage)
        
        print(f"{active_combatant.name} dealt {active_combatant.damage} damage to {victim.name}")

        if victim.is_dead():
            print(f"{victim.name} dies!")
            PQ.remove(victim.name)

        PQ.add(active_combatant)

        print()
        remaining_combatants = PQ.get_alive_combatants()


if game_won(remaining_combatants):
    print("You won!")
else:
    print("You died!")
```
